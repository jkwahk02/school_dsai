# Lesson 7. Univariate Analysis and Descriptive Statistics

1) Measures of Central Tendency

```python
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files

# Upload CSV file
uploaded = files.upload()  # Choose "penguins.csv"

# Load dataset
df = pd.read_csv("penguins.csv")

# Keep numeric columns but exclude 'id' and 'year'
numeric_df = df.select_dtypes(include=[np.number]).drop(columns=['id', 'year'], errors='ignore').dropna()

# Function to calculate mean, median, and mode
def central_tendency(series):
    mean = np.mean(series)
    median = np.median(series)
    mode_result = stats.mode(series, keepdims=True)
    mode_val = mode_result.mode[0]
    mode_count = mode_result.count[0]
    return pd.Series({
        "Mean": round(mean, 2),
        "Median": round(median, 2),
        "Mode": round(mode_val, 2),
        "Mode Count": int(mode_count)
    })

# Apply to all numeric columns
central_results = numeric_df.apply(central_tendency).T  # transpose for cleaner layout

# Show results as a table
print("Central Tendency Measures (Penguins Dataset):")
display(central_results)
```

2, Dispersion

```python
import pandas as pd
import numpy as np
from google.colab import files

# Upload CSV file
uploaded = files.upload()  # Choose "penguins.csv"

# Load dataset
df = pd.read_csv("penguins.csv")

# Keep numeric columns but exclude 'id' and 'year'
numeric_df = df.select_dtypes(include=[np.number]).drop(columns=['id', 'year'], errors='ignore').dropna()

# Function to calculate dispersion measures
def dispersion_measures(series):
    rng = series.max() - series.min()
    std_sample = np.std(series, ddof=1)      # sample standard deviation
    var_sample = np.var(series, ddof=1)      # sample variance
    q1 = np.percentile(series, 25)
    q3 = np.percentile(series, 75)
    iqr = q3 - q1
    return pd.Series({
        "Range": round(rng, 2),
        "Variance (Sample)": round(var_sample, 2),
        "StdDev (Sample)": round(std_sample, 2),
        "IQR": round(iqr, 2)
    })

# Apply to all numeric columns
dispersion_results = numeric_df.apply(dispersion_measures).T  # transpose for cleaner table

# Show results as a table
print("Dispersion Measures (Penguins Dataset):")
display(dispersion_results)
```

3. Distribution Shape

```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis
from google.colab import files

# Upload CSV file from your computer
uploaded = files.upload()

# Get the filename (first uploaded file)
filename = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(filename)

# Keep numeric columns but exclude 'id' and 'year'
numeric_df = df.select_dtypes(include=[np.number]).drop(columns=['id', 'year'], errors='ignore').dropna()

# Function to calculate skewness, kurtosis, and outliers
def shape_and_outliers(series):
    skewness_val = skew(series, bias=False)
    kurt_excess = kurtosis(series, fisher=True, bias=False)  # excess kurtosis (0 = normal)

    # Outlier detection using IQR
    q1, q3 = np.percentile(series, [25, 75])
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = series[(series < lower_bound) | (series > upper_bound)]

    return pd.Series({
        "Skewness": round(skewness_val, 3),
        "Kurtosis (Excess)": round(kurt_excess, 3),
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outlier Count": len(outliers)
    })

# Apply function to all numeric columns
shape_results = numeric_df.apply(shape_and_outliers).T  # transpose to make columns as rows

# Display table
print(f"\nDistribution Shape & Outliers for: {filename}\n")
display(shape_results)
```

6. Box Plot

```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload CSV file
uploaded = files.upload()  # Choose "penguins.csv"
filename = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(filename)

# Drop missing values in body_mass_g
data = df["body_mass_g"].dropna()

# Extract stats
q1 = data.quantile(0.25)
q3 = data.quantile(0.75)
median = data.median()
iqr = q3 - q1
lower_whisker = max(data[data >= q1 - 1.5*iqr].min(), data.min())
upper_whisker = min(data[data <= q3 + 1.5*iqr].max(), data.max())
lower_limit = q1 - 1.5*iqr
upper_limit = q3 + 1.5*iqr

# Create vertical boxplot (wider figure)
plt.figure(figsize=(8,7))  # wider width for more space
plt.boxplot(data, vert=True, patch_artist=True,
            boxprops=dict(facecolor="lightblue", color="blue"),
            medianprops=dict(color="red", linewidth=2),
            whiskerprops=dict(color="blue"),
            capprops=dict(color="blue"),
            flierprops=dict(markerfacecolor="orange", marker="o", markersize=6, alpha=0.6))

plt.title("Box Plot of Penguin Body Mass (g) with Labels", fontsize=14)
plt.ylabel("Body Mass (g)")

# Adjust y-axis to include outlier cutoff limits
plt.ylim(lower_limit - 200, upper_limit + 200)

# Annotate quartiles, whiskers, IQR, and limits
plt.text(1.15, median, f"Median = {median:.0f}", color="red", va="center")
plt.text(1.15, q1, f"Q1 = {q1:.0f}", color="blue", va="center")
plt.text(1.15, q3, f"Q3 = {q3:.0f}", color="blue", va="center")
plt.text(1.15, lower_whisker, f"Whisker Min = {lower_whisker:.0f}", color="purple", va="center")
plt.text(1.15, upper_whisker, f"Whisker Max = {upper_whisker:.0f}", color="purple", va="center")
plt.text(1.15, lower_limit, f"Outlier Limit Low = {lower_limit:.0f}", color="orange", va="center")
plt.text(1.15, upper_limit, f"Outlier Limit High = {upper_limit:.0f}", color="orange", va="center")
plt.text(0.7, (q1+q3)/2, f"IQR = {iqr:.0f}", color="green", rotation=90, va="center")

plt.show()
```

```python
import pandas as pd
from google.colab import files

# Upload CSV
uploaded = files.upload()  # Choose "penguins.csv"
filename = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(filename)

# Choose numeric column (body_mass_g)
data = df["body_mass_g"].dropna().astype(int)

def stem_and_leaf(data, scale=100):
    stems = {}
    for val in sorted(data):
        stem, leaf = divmod(val, scale)
        stems.setdefault(stem, []).append(str(leaf).zfill(len(str(scale))-1))
    # Print in classic stem-leaf style
    for stem, leaves in stems.items():
        print(f"{stem} | {' '.join(leaves)}")

print("Stem-and-Leaf Plot for body_mass_g:\n")
stem_and_leaf(data, scale=100)  # stem = hundreds, leaf = last two digits
```

2. Chi-square Test of Independence

```python
import pandas as pd
import scipy.stats as stats
from google.colab import files

uploaded = files.upload()  # Choose "penguins.csv"
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# Contingency table
contingency_table = pd.crosstab(df["species"], df["island"])

# Chi-square test
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

# Reformat expected frequencies into a DataFrame
expected_df = pd.DataFrame(expected,
                           index=contingency_table.index,
                           columns=contingency_table.columns)

print("Chi-square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("p-value:", p)
print("\nObserved Frequencies:\n", contingency_table)
print("\nExpected Frequencies (rounded):\n", expected_df.round(2))
```

3. Fisher's Exact Test

```python
import pandas as pd
import scipy.stats as stats
from google.colab import files

# Upload CSV
uploaded = files.upload()  # Choose "penguins.csv"
filename = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(filename)

# Example: Fisher’s Exact Test (Adelie vs Gentoo × Sex)
# Subset data to only include Adelie and Gentoo
subset = df[df["species"].isin(["Adelie", "Gentoo"])].dropna(subset=["sex"])

# Build 2x2 contingency table
contingency_table = pd.crosstab(subset["species"], subset["sex"])
print("Observed Frequencies:\n", contingency_table)

# Run Fisher’s Exact Test
oddsratio, p_value = stats.fisher_exact(contingency_table)
print("\nOdds Ratio:", round(oddsratio, 3))
print("Fisher’s Exact Test p-value:", round(p_value, 4))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats
from sklearn.linear_model import LinearRegression

# Upload and load data
from google.colab import files
uploaded = files.upload()  # Choose "penguins.csv"
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# Clean data - remove missing values for relevant columns
df_clean = df.dropna(subset=['flipper_length_mm', 'body_mass_g', 'species'])

plt.figure(figsize=(10, 8))

# Scatter plot with regression lines for each species
ax = sns.scatterplot(
    data=df_clean, x="flipper_length_mm", y="body_mass_g",
    hue="species", palette="Set2", s=60, alpha=0.7
)

species_list = df_clean['species'].unique()
colors = ['#fc8d62', '#66c2a5', '#8da0cb']  # Set2 palette colors

for i, species in enumerate(species_list):
    # Species-specific data
    species_data = df_clean[df_clean['species'] == species]
    x_vals = species_data['flipper_length_mm']
    y_vals = species_data['body_mass_g']

    # Calculate correlation
    correlation, p_value = stats.pearsonr(x_vals, y_vals)

    # Fit regression line
    reg = LinearRegression().fit(x_vals.values.reshape(-1, 1), y_vals)
    slope = reg.coef_[0]
    intercept = reg.intercept_

    # Draw regression line
    x_line = np.linspace(x_vals.min(), x_vals.max(), 100)
    y_line = slope * x_line + intercept
    plt.plot(x_line, y_line, color=colors[i], linewidth=2, alpha=0.8)

    # Annotate plot with correlation and regression equation
    text_positions = [(0.02, 0.95), (0.02, 0.88), (0.02, 0.81)]  # Avoid overlap
    plt.text(
        text_positions[i][0], text_positions[i][1],
        f'{species}: r = {correlation:.3f}, p = {p_value:.3f}\ny = {slope:.2f}x + {intercept:.2f}',
        transform=ax.transAxes, fontsize=10,
        bbox=dict(boxstyle="round,pad=0.3", facecolor=colors[i], alpha=0.3),
        verticalalignment='top'
    )

# Overall correlation for all data
overall_corr, overall_p = stats.pearsonr(df_clean['flipper_length_mm'], df_clean['body_mass_g'])
plt.text(
    0.02, 0.15,
    f'Overall: r = {overall_corr:.3f}, p = {overall_p:.3f}',
    transform=ax.transAxes, fontsize=11, fontweight='bold',
    bbox=dict(boxstyle="round,pad=0.3", facecolor='lightgray', alpha=0.5),
    verticalalignment='top'
)

plt.title("Scatter Plot: Flipper Length vs Body Mass\nwith Correlation Coefficients and Regression Lines",
          fontsize=14, fontweight='bold', pad=20)
plt.xlabel("Flipper Length (mm)", fontsize=12)
plt.ylabel("Body Mass (g)", fontsize=12)
plt.legend(title='Species', title_fontsize=11, fontsize=10, loc='lower right')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

def load_and_prepare_data():
    uploaded = files.upload()
    filename = list(uploaded.keys())[0]
    df = pd.read_csv(filename)
    df = df.drop(columns=[c for c in ['species', 'island', 'sex', 'year', 'id'] if c in df.columns])
    df = df.dropna()
    X = df.select_dtypes(include=['number'])
    return df, X

def standardize_data(X):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return X_scaled, scaler

df, X = load_and_prepare_data()
X_scaled, scaler = standardize_data(X)
```

```python

def plot_elbow_and_silhouette(X_scaled, k_range=range(2,10)):
    inertia = []
    silhouette_scores = []
    for k in k_range:
        kmeans = KMeans(n_clusters=k, random_state=42)
        labels = kmeans.fit_predict(X_scaled)
        inertia.append(kmeans.inertia_)
        silhouette_scores.append(silhouette_score(X_scaled, labels))
    plt.figure(figsize=(8,4))
    plt.plot(k_range, inertia, 'bo-', label="WSS (Inertia)")
    plt.xlabel("Number of Clusters (k)")
    plt.ylabel("Within-Cluster Sum of Squares")
    plt.title("Elbow Method for Optimal k")
    plt.legend()
    plt.show()

    plt.figure(figsize=(8,4))
    plt.plot(k_range, silhouette_scores, 'go-', label="Silhouette Score")
    plt.xlabel("Number of Clusters (k)")
    plt.ylabel("Silhouette Score")
    plt.title("Silhouette Analysis for Optimal k")
    plt.legend()
    plt.show()
    return inertia, silhouette_scores

plot_elbow_and_silhouette(X_scaled)
```

```python

def perform_kmeans(X_scaled, original_df, scaler, n_clusters=3):
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    clusters = kmeans.fit_predict(X_scaled)
    original_df['Cluster'] = clusters
    wss = kmeans.inertia_
    means = np.mean(X_scaled, axis=0)
    bss = sum(np.sum((kmeans.cluster_centers_ - means) ** 2, axis=1) * np.bincount(clusters))
    silhouette_avg = silhouette_score(X_scaled, clusters)
    return kmeans, original_df, wss, bss, silhouette_avg

def display_results(kmeans, df, scaler):
    print(f"\nWithin-Cluster Sum of Squares (WSS): {kmeans.inertia_:.2f}")
    # Select only numeric columns that were used for fitting the scaler
    numeric_cols_for_scaling = df.select_dtypes(include=['number']).columns.tolist()
    if 'Cluster' in numeric_cols_for_scaling:
      numeric_cols_for_scaling.remove('Cluster')

    means = np.mean(scaler.transform(df[numeric_cols_for_scaling]), axis=0)
    bss = sum(np.sum((kmeans.cluster_centers_ - means) ** 2, axis=1) * np.bincount(df['Cluster']))
    print(f"Between-Cluster Sum of Squares (BSS): {bss:.2f}")
    silhouette_avg = silhouette_score(scaler.transform(df[numeric_cols_for_scaling]), df['Cluster'])
    print(f"Silhouette Score: {silhouette_avg:.3f}")

    centroids = scaler.inverse_transform(kmeans.cluster_centers_)
    centroid_df = pd.DataFrame(centroids, columns=numeric_cols_for_scaling) # Use correct columns
    print("\nCluster Centroids (Original Scale):")
    display(centroid_df.style.background_gradient(cmap='viridis'))

    stats_df = df.groupby('Cluster').agg(['count', 'mean', 'std', 'min', 'max'])
    print("\nCluster Descriptive Statistics:")
    display(stats_df.style.background_gradient(cmap='coolwarm'))

    print("\nCluster Size Distribution:")
    print(df['Cluster'].value_counts())

kmeans_model, updated_df, wss, bss, silhouette_avg = perform_kmeans(X_scaled, df, scaler)
display_results(kmeans_model, updated_df, scaler)
```

```python


def visualize_clusters_pairplot(df, numeric_cols, cluster_col='Cluster'):
    """
    Visualize clusters using pairwise scatterplots colored by cluster assignment.
    """
    sns.pairplot(df, vars=numeric_cols, hue=cluster_col, palette='Set2', diag_kind='kde', corner=True)
    plt.suptitle("Cluster Visualization with Pairwise Scatterplots", y=1.02)
    plt.show()

def plot_dendrogram(X_scaled):
    linked = linkage(X_scaled, method='ward')
    plt.figure(figsize=(12,6))
    dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
    plt.title("Hierarchical Clustering Dendrogram")
    plt.xlabel("Sample Index")
    plt.ylabel("Euclidean Distance")
    plt.show()

visualize_clusters_pairplot(updated_df, list(X.columns), cluster_col='Cluster')
plot_dendrogram(X_scaled)
```

```python
# ======================================
# Descriptive Analytics Toolkit (Weekly) + Advanced KPIs
# ======================================

# 0) Setup and CSV Upload
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from io import StringIO
from google.colab import files

pd.set_option("display.max_columns", 100)

print("Please upload your CSV file...")
uploaded = files.upload()
filename = next(iter(uploaded))
df = pd.read_csv(StringIO(uploaded[filename].decode('utf-8')))
print(f"File '{filename}' loaded. Shape: {df.shape}")
display(df.head())

# 1) Column Detection Helpers
def find_col(cols, candidates):
    """Return first column that matches any candidate (substring match, case-insensitive)."""
    cols_l = [c.lower() for c in cols]
    for cand in candidates:
        cand_l = cand.lower()
        for i, c in enumerate(cols_l):
            if cand_l in c:
                return cols[i]
    return None

def guess_datetime_col(frame):
    """Try to find a datetime-like column by attempting to parse each column; pick the one with most parses."""
    best_col, best_valid = None, 0
    for c in frame.columns:
        s = pd.to_datetime(frame[c], errors="coerce", infer_datetime_format=True)
        valid = s.notna().sum()
        if valid > best_valid and valid > 0:
            best_col, best_valid = c, valid
    return best_col

cols = df.columns.tolist()

# Synonym dictionary (broad)
SYN = {
    "date": ["order_date","date","purchase","created_at","timestamp","order_ts","datetime","time"],
    "delivery_date": ["delivery_date","delivered_at","ship_date","shipping_date","fulfillment_date"],
    "price": ["discounted_price","selling_price","unit_price","price","sale_price","net_price"],
    "actual_price": ["actual_price","original_price","mrp","list_price","base_price","regular_price"],
    "quantity": ["quantity","qty","units","order_item_quantity","unit_sold","units_sold"],
    "revenue": ["revenue","sales","amount","order_value","gross_sales","net_sales","gmv"],
    "cost": ["cost","cogs","cost_of_goods_sold"],
    "profit": ["profit","margin_amount","gross_profit","net_profit"],
    "discount_value": ["discount","discount_amount","discount_value","markdown"],
    "refund": ["refund","returns","refund_amount","return_amount","cancellation_amount"],
    "category": ["category","product_category","main_category","dept","department"],
    "product": ["product","product_id","asin","sku","title","item_name","product_title","item"],
    "channel": ["channel","marketplace","platform","source","utm_source","store","region"],
    "rating": ["rating","review_rating","stars","product_rating","avg_rating"],
    "order_id": ["order_id","order","txn_id","transaction_id","order_number"],
    "customer_id": ["customer_id","user_id","buyer_id","client_id","member_id"]
}


# Build initial mapping
mapping = {}
for key, cand in SYN.items():
    mapping[key] = find_col(cols, cand)

# Date fallback: guess if needed
if mapping["date"] is None:
    mapping["date"] = guess_datetime_col(df)

# Apply manual overrides if provided - keep this for manual testing
OVERRIDE = {
    # Example: "price": "unit_price", "quantity": "qty", "date": "created_at"
    # Leave empty to use inference.
}
for k, v in OVERRIDE.items():
    mapping[k] = v

print("Inferred/Overridden column mapping:")
print(mapping)

# 2) Build Core Value Metric and Weekly Period
# Core value logic: prefer revenue; else price*qty; else price
value_col = None
if mapping["revenue"] and mapping["revenue"] in df.columns:
    value_col = mapping["revenue"]
elif mapping["price"] and mapping["price"] in df.columns and mapping["quantity"] and mapping["quantity"] in df.columns:
    df["_value"] = pd.to_numeric(df[mapping["price"]], errors="coerce") * pd.to_numeric(df[mapping["quantity"]], errors="coerce")
    value_col = "_value"
else:
    value_col = mapping["price"] # fallback; many KPIs still work

# Parse dates + weekly bins
if mapping["date"] and mapping["date"] in df.columns:
    df[mapping["date"]] = pd.to_datetime(df[mapping["date"]], errors="coerce", infer_datetime_format=True)
    week_period = df[mapping["date"]].dt.to_period(CONFIG["WEEK_FREQ"])
    df["_week"] = week_period.astype(str)  # e.g., '2023-01-02/2023-01-08'
    # Use elementwise start_time for pandas compatibility
    df["_week_start"] = week_period.apply(lambda p: p.start_time.date() if pd.notnull(p) else np.nan)
else:
    df["_week"] = "All"
    df["_week_start"] = "All"

# 3) Utility Functions and Core KPI Calculator
def safe_sum(series): return float(pd.to_numeric(series, errors="coerce").sum()) if series is not None and not series.empty else np.nan
def safe_mean(series):
    s = pd.to_numeric(series, errors="coerce")
    return float(s.mean()) if s.notna().any() else np.nan

def compute_kpis(frame: pd.DataFrame) -> dict:
    k = {}
    # Sales & units
    if value_col and value_col in frame:
        k["Total Sales"] = safe_sum(frame[value_col])
        k["Avg Sales/Row"] = safe_mean(frame[value_col])
    if mapping["quantity"] and mapping["quantity"] in frame:
        k["Total Units"] = safe_sum(frame[mapping["quantity"]])
        if mapping["order_id"] and mapping["order_id"] in frame:
            n_orders = frame[mapping["order_id"]].nunique()
            k["UPT (Units/Txn)"] = (k["Total Units"]/n_orders) if n_orders else np.nan
    # Price & margins
    if mapping["price"] and mapping["price"] in frame:
        k["ASP (Avg Selling Price)"] = safe_mean(frame[mapping["price"]])
    if mapping["profit"] and mapping["profit"] in frame:
        k["Total Profit"] = safe_sum(frame[mapping["profit"]])
    if mapping["profit"] and mapping["profit"] in frame and value_col and value_col in frame:
        sales = k.get("Total Sales", safe_sum(frame[value_col]))
        profit = safe_sum(frame[mapping["profit"]])
        k["Gross Margin %"] = (profit / sales * 100) if sales else np.nan
    # Discounts
    if mapping["actual_price"] and mapping["actual_price"] in frame and mapping["price"] and mapping["price"] in frame:
        ap = pd.to_numeric(frame[mapping["actual_price"]], errors="coerce")
        sp = pd.to_numeric(frame[mapping["price"]], errors="coerce")
        pct_off = (ap - sp) / ap
        k["Average Discount %"] = float(pct_off.replace([np.inf, -np.inf], np.nan).mean() * 100) if pct_off.notna().any() else np.nan
        k["Total Discount"] = float((ap - sp).clip(lower=0).sum())
        if value_col and value_col in frame:
            total_sales = safe_sum(frame[value_col])
            k["Discount % of Sales"] = (k["Total Discount"]/total_sales*100) if total_sales else np.nan
        # Margin proxy if cost not available
        retained = sp / ap
        k["Margin Proxy %"] = float(retained.replace([np.inf, -np.inf], np.nan).mean()*100) if retained.notna().any() else np.nan
    elif mapping["discount_value"] and mapping["discount_value"] in frame and value_col and value_col in frame:
        total_disc = safe_sum(frame[mapping["discount_value"]])
        total_sales = safe_sum(frame[value_col])
        k["Total Discount"] = total_disc
        k["Discount % of Sales"] = (total_disc / total_sales * 100) if total_sales else np.nan
    # Refunds
    if mapping["refund"] and mapping["refund"] in frame and value_col and value_col in frame:
        refunds = safe_sum(frame[mapping["refund"]])
        sales = safe_sum(frame[value_col])
        k["Refunds"] = refunds
        k["Refund Rate %"] = (refunds/sales*100) if sales else np.nan
    # Breadth & customers
    if mapping["product"] and mapping["product"] in frame: k["Active SKUs"] = int(frame[mapping["product"]].nunique())
    if mapping["category"] and mapping["category"] in frame: k["Active Categories"] = int(frame[mapping["category"]].nunique())
    if mapping["rating"] and mapping["rating"] in frame: k["Avg Rating"] = safe_mean(frame[mapping["rating"]])
    if mapping["order_id"] and mapping["order_id"] in frame: k["Orders"] = int(frame[mapping["order_id"]].nunique())
    if mapping["customer_id"] and mapping["customer_id"] in frame: k["Active Customers"] = int(frame[mapping["customer_id"]].nunique())
    # AOV
    if mapping["order_id"] and mapping["order_id"] in frame and value_col and value_col in frame:
        tmp = frame[[mapping["order_id"], value_col]].copy()
        tmp[value_col] = to_num(tmp[value_col])
        order_sales = tmp.groupby(mapping["order_id"])[value_col].sum()
        k["AOV (Avg Order Value)"] = float(order_sales.mean()) if len(order_sales) else np.nan
    # Channel share
    if mapping["channel"] and mapping["channel"] in frame and value_col and value_col in frame:
        ch = frame.groupby(mapping["channel"])[value_col].sum().sort_values(ascending=False)
        total = ch.sum()
        if total and not ch.empty:
            k["Top Channel"] = str(ch.index[0])
            k["Top Channel Share %"] = float(ch.iloc[0] / total * 100)
    return k

def kpi_row(frame, label):
    row = {"Label": label}
    row.update(compute_kpis(frame))
    return row

def format_kpis(table: pd.DataFrame) -> pd.DataFrame:
    tbl = table.copy()
    pct_cols = [c for c in tbl.columns if c.endswith("%")]
    for c in pct_cols:
        tbl[c] = tbl[c].map(lambda x: f"{x:.2f}%" if pd.notna(x) else "")
    num_cols = [c for c in tbl.columns if c not in ["Label"] + pct_cols]
    for c in num_cols:
        if c in tbl:
            tbl[c] = tbl[c].map(lambda x: f"{x:,.2f}" if pd.notna(x) else "")
    return tbl

# -----------------------------
# 5) Overall & Weekly KPI tables
# -----------------------------
overall_kpis = pd.DataFrame([kpi_row(df, "Overall")])
display(format_kpis(overall_kpis.copy()))

if "_week_start" in df.columns and df["_week_start"].notna().any():
    weekly_rows = [kpi_row(g, wk) for wk, g in df.groupby("_week_start")]
    weekly_kpis = pd.DataFrame(weekly_rows).sort_values("Label")

    cols_show = [c for c in [
        "Label","Total Sales","Orders","AOV (Avg Order Value)","Total Units","UPT (Units/Txn)",
        "ASP (Avg Selling Price)","Total Profit","Gross Margin %",
        "Total Discount","Average Discount %","Discount % of Sales","Margin Proxy %",
        "Refunds","Refund Rate %",
        "Active SKUs","Active Categories","Avg Rating","Active Customers",
        "Top Channel","Top Channel Share %"
    ] if c in weekly_kpis.columns]
    display(format_kpis(weekly_kpis[cols_show].copy()).head(25))

    # -----------------------------
    # 6) Trend (WoW) on primary metric
    # -----------------------------
    metric = "Total Sales" if "Total Sales" in weekly_kpis.columns else (
        next((c for c in weekly_kpis.columns if c not in ["Label"]), None)
    )

    if metric:
        trend = weekly_kpis[["Label", metric]].copy().rename(columns={"Label": "WeekStart"})
        trend["Prev"] = trend[metric].shift(1)
        trend["Abs Change"] = trend[metric] - trend["Prev"]
        trend["Percent Change"] = np.where(trend["Prev"].notna(),
                                           (trend[metric] - trend["Prev"]) / trend["Prev"] * 100,
                                           np.nan)
        display(trend.head(15))

        plt.figure()
        plt.plot(trend["WeekStart"], trend[metric], marker="o")
        plt.title(f"{metric} by Week")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

        # -----------------------------
        # 7) Benchmarking & Actionable Notes
        # -----------------------------
        baseline = trend[metric].median()
        target = baseline * 1.05
        trend["Target"] = target
        trend["On Track?"] = np.where(trend[metric] >= target, "Yes", "No")
        display(trend.tail(15))

        notes = []
        thr = CONFIG["WOW_ALERT_THRESH"]
        for _, r in trend.iterrows():
            msg = []
            if pd.notna(r["Percent Change"]):
                if r["Percent Change"] < -thr: msg.append(f"{metric} down more than {thr}% WoW; investigate.")
                elif r["Percent Change"] >  thr: msg.append(f"{metric} up more than {thr}% WoW; reinforce drivers.")
            notes.append({"WeekStart": r["WeekStart"], "Notes": " | ".join(msg) if msg else "No major changes."})
        action_notes = pd.DataFrame(notes)
        display(action_notes.head(15))
    else:
        print("Could not determine a metric for weekly trend analysis.")
else:
    print("Weekly analysis skipped as date_col was not inferred or is all missing values.")


# -----------------------------
# 8) Stakeholder Views (Channel & Category)
# -----------------------------
if mapping["channel"] and mapping["channel"] in df.columns:
    ch_rows = [kpi_row(g, ch) for ch, g in df.groupby(mapping["channel"])]
    if ch_rows:
        ch_kpis = pd.DataFrame(ch_rows).sort_values("Label")
        display(format_kpis(ch_kpis.copy()))
        if value_col and value_col in df:
            ch_share = df.groupby(mapping["channel"])[value_col].sum().sort_values(ascending=False)
            if ch_share.sum() > 0:
                 ch_share_pct = (ch_share / ch_share.sum() * 100).round(2)
                 display(pd.DataFrame({"Channel": ch_share_pct.index, "Sales Share %": ch_share_pct.values}))
    else:
        print("No data to group by channel.")
else:
    print("Channel column not inferred for analysis.")

if mapping["category"] and mapping["category"] in df.columns:
    cat_rows = [kpi_row(g, c) for c, g in df.groupby(mapping["category"])]
    if cat_rows:
        cat_kpis = pd.DataFrame(cat_rows)
        if "Total Sales" in cat_kpis:
            cat_kpis = cat_kpis.sort_values("Total Sales", ascending=False).head(CONFIG["TOP_CATEGORIES"])
        display(format_kpis(cat_kpis.copy()))
    else:
        print("No data to group by category.")
else:
    print("Category column not inferred for analysis.")


# -----------------------------
# 9) Advanced KPIs (generic)
# -----------------------------
print("\n--- Advanced KPIs ---")

# 9.1 WASP
wasp = np.nan
if mapping["price"] and mapping["price"] in df.columns and (mapping["quantity"] in df.columns):
    price_num = to_num(df[mapping["price"]])
    qty_num = to_num(df[mapping["quantity"]])
    total_units = qty_num.sum()
    if total_units > 0:
        wasp = (price_num * qty_num).sum() / total_units
print("Weighted Avg Selling Price (WASP):", wasp)

# 9.2 Elasticity proxy (Discount% vs Units corr)
elasticity_corr = np.nan
if mapping["actual_price"] and mapping["actual_price"] in df.columns and mapping["price"] and mapping["price"] in df.columns and (mapping["quantity"] in df.columns):
    ap = to_num(df[mapping["actual_price"]])
    sp = to_num(df[mapping["price"]])
    qty_num = to_num(df[mapping["quantity"]])
    # Handle division by zero/inf/nan explicitly
    disc_pct = (ap - sp) / ap
    disc_pct = disc_pct.replace([np.inf, -np.inf], np.nan)

    valid_data = pd.concat([disc_pct.rename("disc_pct"), qty_num.rename("units")], axis=1).dropna()

    if len(valid_data) > 1:
        # Calculate correlation only on valid data
        elasticity_corr = valid_data["disc_pct"].corr(valid_data["units"])
print("Discount % vs Units correlation (elasticity proxy):", elasticity_corr)


# 9.3 Rating KPIs
rating_summary = {}
if mapping["rating"] and mapping["rating"] in df.columns:
    r = to_num(df[mapping["rating"]])
    # Handle cases where rating is all NaN
    if r.notna().sum() > 0:
      dist = r.dropna().round().value_counts(normalize=True).sort_index() * 100
      rating_summary["Distribution % (by star)"] = dist

      if value_col and value_col in df:
          value_num = to_num(df[value_col])
          if value_num.notna().sum() > 0:
             wr = (r * value_num).sum() / value_num.sum()
             rating_summary["Weighted Avg Rating (by sales)"] = wr
      elif mapping["quantity"] and mapping["quantity"] in df.columns:
          qty_num = to_num(df[mapping["quantity"]])
          if qty_num.notna().sum() > 0:
             wr = (r * qty_num).sum() / qty_num.sum()
             rating_summary["Weighted Avg Rating (by units)"] = wr

      if mapping["product"] and mapping["product"] in df.columns:
          # Calculate average rating per product, dropping products with no rating data
          prod_avg = df.groupby(mapping["product"])[mapping["rating"]].mean(numeric_only=True).dropna()
          if len(prod_avg) > 0:
            rating_summary["Products <3.5 Stars (%)"] = (prod_avg < 3.5).mean() * 100
      print("Rating KPIs:")
      for k, v in rating_summary.items():
          if isinstance(v, pd.Series):
              display(v.rename_axis("Stars").reset_index(name="Share %"))
          else:
              print(k, ":", v)
    else:
        print("Rating column found but contains no valid data.")

else:
    print("No rating column inferred or found.")




# 9.5 Long-tail share (SKU concentration)
long_tail_share = np.nan
if mapping["product"] and mapping["product"] in df.columns and value_col and value_col in df:
    # Ensure value_col is numeric
    df[value_col] = to_num(df[value_col])
    # Drop NA for relevant columns before grouping
    valid_df_for_longtail = df.dropna(subset=[mapping["product"], value_col])
    if len(valid_df_for_longtail) > 0:
        sku_sales = valid_df_for_longtail.groupby(mapping["product"])[value_col].sum().sort_values(ascending=False)
        if sku_sales.sum() > 0:
            cum_share = sku_sales.cumsum() / sku_sales.sum()
            # Find the point where cumulative share exceeds 80%
            top_80_index = cum_share[cum_share <= 0.8].index
            if not top_80_index.empty:
                # Calculate share of items *after* the 80% cutoff
                long_tail_share = 1 - cum_share[top_80_index[-1]]
            else:
                # If no SKU gets to 80%, the long tail is the entire dataset
                long_tail_share = 1.0
        else:
            long_tail_share = 0.0 # No sales
    else:
        long_tail_share = np.nan # No valid data for calculation

print("Long-tail Share (%):", long_tail_share * 100 if pd.notna(long_tail_share) else np.nan)


# 9.6 Category Performance Index
cat_perf_index = pd.DataFrame()
if mapping["category"] and mapping["category"] in df.columns and value_col and value_col in df:
    # Ensure value_col is numeric
    df[value_col] = to_num(df[value_col])
    # Drop NA for relevant columns before grouping
    valid_df_for_cat_perf = df.dropna(subset=[mapping["category"], value_col])
    if len(valid_df_for_cat_perf) > 0:
        cat_sales = valid_df_for_cat_perf.groupby(mapping["category"])[value_col].sum()
        if cat_sales.mean() > 0: # Avoid division by zero if mean is 0
           cat_perf_index = (cat_sales / cat_sales.mean() * 100).sort_values(ascending=False).rename("Category Performance Index").reset_index()
           display(cat_perf_index.head(20))
        else:
           print("Category Performance Index skipped: Average category sales is zero.")
    else:
        print("Category Performance Index skipped: No valid data for calculation.")

else:
    print("Category Performance Index skipped (need category and value columns).")


# 9.7 Weekly MA & Volatility
weekly_ma = pd.DataFrame()
if mapping["date"] and mapping["date"] in df.columns and value_col and value_col in df:
    # Ensure date column is datetime and value column is numeric
    df[mapping["date"]] = pd.to_datetime(df[mapping["date"]], errors="coerce")
    df[value_col] = to_num(df[value_col])

    # Drop rows where date or value is missing for time series analysis
    valid_df_for_ma = df.dropna(subset=[mapping["date"], value_col]).copy()

    if len(valid_df_for_ma) > 0:
        # Group by the inferred date column after ensuring it's datetime
        wk_sales = valid_df_for_ma.set_index(mapping["date"]).groupby(pd.Grouper(freq=CONFIG["WEEK_FREQ"]))[value_col].sum().rename("sales").reset_index()

        if len(wk_sales) > 1: # Need at least 2 weeks for MA/Volatility
            wk_sales["MA_4w"] = wk_sales["sales"].rolling(window=4, min_periods=1).mean()
            wk_sales["Volatility_σ_8w"] = wk_sales["sales"].rolling(window=8, min_periods=2).std()
            weekly_ma = wk_sales.copy()
            display(weekly_ma.head(12))

            plt.figure()
            plt.plot(weekly_ma[mapping["date"]], weekly_ma["sales"], marker="o", label="Sales")
            plt.plot(weekly_ma[mapping["date"]], weekly_ma["MA_4w"], label="4w MA")
            plt.xticks(rotation=45, ha="right")
            plt.legend()
            plt.title("Weekly Sales and 4-week Moving Average")
            plt.tight_layout()
            plt.show()
        else:
            print("Weekly MA & Volatility skipped: Need at least 2 weeks of data with sales.")
    else:
         print("Weekly MA & Volatility skipped: No valid date and sales data found.")

else:
    print("Weekly MA & Volatility skipped (need date and value columns).")


# 9.8 Category Mix Shift (WoW)
mix_change = pd.DataFrame()
if mapping["date"] and mapping["date"] in df.columns and mapping["category"] and mapping["category"] in df.columns and value_col and value_col in df:
    # Ensure date and value columns are correct types
    df[mapping["date"]] = pd.to_datetime(df[mapping["date"]], errors="coerce")
    df[value_col] = to_num(df[value_col])

    # Drop NA for relevant columns
    valid_df_for_mix = df.dropna(subset=[mapping["date"], mapping["category"], value_col]).copy()

    if len(valid_df_for_mix) > 0:
        # Group by week and category
        wkcat = valid_df_for_mix.set_index(mapping["date"]).groupby([pd.Grouper(freq=CONFIG["WEEK_FREQ"]), mapping["category"]])[value_col].sum().reset_index()

        # Calculate total sales by week
        total_by_week = wkcat.groupby(mapping["date"])[value_col].transform("sum")

        # Avoid division by zero if a week has zero total sales
        wkcat["Share%"] = np.where(total_by_week > 0, wkcat[value_col] / total_by_week * 100, 0)

        # Pivot to get categories as columns and weeks as index
        mix = wkcat.pivot(index=mapping["date"], columns=mapping["category"], values="Share%").sort_index()

        # Calculate week-over-week change
        mix_change = mix.diff(periods=1)
        display(mix_change.head(10))
    else:
        print("Category Mix Shift skipped: No valid data found for calculation.")

else:
    print("Category Mix Shift skipped (need date, category, and value columns).")


# 9.9 Correlations
corr_sales_rating = np.nan
corr_price_rating = np.nan
if mapping["rating"] and mapping["rating"] in df.columns:
    r = to_num(df[mapping["rating"]])
    if r.notna().sum() > 1: # Need at least 2 non-NaN values for correlation
        if value_col and value_col in df:
            value_num = to_num(df[value_col])
            valid_corr_s = pd.concat([r.rename("rating"), value_num.rename("sales")], axis=1).dropna()
            if len(valid_corr_s) > 1:
                 corr_sales_rating = valid_corr_s.corr().iloc[0,1]
            else:
                 print("Correlation (Rating vs Sales) skipped: Not enough overlapping non-missing data.")

        if mapping["price"] and mapping["price"] in df.columns:
            price_num = to_num(df[mapping["price"]])
            valid_corr_p = pd.concat([r.rename("rating"), price_num.rename("price")], axis=1).dropna()
            if len(valid_corr_p) > 1:
                corr_price_rating = valid_corr_p.corr().iloc[0,1]
            else:
                print("Correlation (Rating vs Price) skipped: Not enough overlapping non-missing data.")
    else:
        print("Correlations skipped: Rating column contains less than 2 valid values.")

else:
    print("Correlations skipped (need rating column).")

print("Correlation (Rating vs Sales):", corr_sales_rating)
print("Correlation (Rating vs Price):", corr_price_rating)


# -----------------------------
# 10) Export to Excel (multi-sheet)
# -----------------------------
out = "generic_descriptive_analytics_weekly.xlsx"
with pd.ExcelWriter(out) as xw:
    overall_kpis.to_excel(xw, index=False, sheet_name="Overall_KPIs")
    if 'weekly_kpis' in locals() and len(weekly_kpis):
        weekly_kpis.to_excel(xw, index=False, sheet_name="Weekly_KPIs")
    if 'trend' in locals() and len(trend):
        trend.to_excel(xw, index=False, sheet_name="Trend_WoW")
    if 'ch_kpis' in locals() and len(ch_kpis):
        ch_kpis.to_excel(xw, index=False, sheet_name="Channel_KPIs")
    if 'cat_kpis' in locals() and len(cat_kpis):
        cat_kpis.to_excel(xw, index=False, sheet_name="Category_KPIs")

    # Advanced Summary sheet
    advanced_summary_data = {
        "WASP":[wasp],
        "Elasticity Corr":[elasticity_corr],
        "LongTail%":[long_tail_share*100 if pd.notna(long_tail_share) else np.nan],
        "Corr_Rating_Sales":[corr_sales_rating],
        "Corr_Rating_Price":[corr_price_rating],
    }
    pd.DataFrame(advanced_summary_data).to_excel(xw, index=False, sheet_name="Advanced_Summary")

    if isinstance(rating_summary.get("Distribution % (by star)"), pd.Series):
        rating_summary["Distribution % (by star)"].rename_axis("Stars").reset_index(name="Share %").to_excel(xw, index=False, sheet_name="Rating_Distribution")
    if 'cat_perf_index' in locals() and len(cat_perf_index):
        cat_perf_index.to_excel(xw, index=False, sheet_name="Category_Perf_Index")
    if 'weekly_ma' in locals() and len(weekly_ma):
        # Ensure the date column used as index for plot is also in the Excel
        weekly_ma.to_excel(xw, index=False, sheet_name="Weekly_MA_Volatility")
    if 'mix_change' in locals() and len(mix_change):
         # Reset index to include the date column in Excel
        mix_change.reset_index().to_excel(xw, index=False, sheet_name="Category_Mix_Shift")
    if len(fulfillment_summary):
        pd.DataFrame(list(fulfillment_summary.items()), columns=["Metric","Value"]).to_excel(xw, index=False, sheet_name="Fulfillment")


files.download(out)
```

1. KPI Computation Utilities

```python
def safe_sum(s):
    try: return float(np.nansum(s))
    except: return np.nan

def safe_mean(s):
    try: return float(np.nanmean(s))
    except: return np.nan

def compute_kpis(frame: pd.DataFrame) -> dict:
    k = {}
    # Sales and units
    if value_col in frame:
        k["Total Sales"] = safe_sum(frame[value_col])
        k["Avg Sales/Row"] = safe_mean(frame[value_col])
    if qty_col in frame:
        k["Total Units"] = safe_sum(frame[qty_col])
        if order_id_col in frame:
            orders_n = frame[order_id_col].nunique()
            k["UPT (Units/Txn)"] = (safe_sum(frame[qty_col]) / orders_n) if orders_n else np.nan
    # Pricing/margins
    if price_col in frame:
        k["ASP (Avg Selling Price)"] = safe_mean(frame[price_col])
    if profit_col in frame:
        k["Total Profit"] = safe_sum(frame[profit_col])
    if profit_col in frame and value_col in frame:
        total_sales = safe_sum(frame[value_col])
        total_profit = safe_sum(frame[profit_col])
        k["Gross Margin %"] = (total_profit / total_sales * 100) if total_sales else np.nan
    # Discounts/returns
    if discount_col in frame:
        total_disc = safe_sum(frame[discount_col])
        k["Total Discount"] = total_disc
        if value_col in frame:
            total_sales = safe_sum(frame[value_col])
            k["Discount % of Sales"] = (total_disc / total_sales * 100) if total_sales else np.nan
    if refund_col in frame and value_col in frame:
        refunds = safe_sum(frame[refund_col])
        total_sales = safe_sum(frame[value_col])
        k["Refunds"] = refunds
        k["Refund Rate %"] = (refunds / total_sales * 100) if total_sales else np.nan
    # Breadth and customers
    if product_col in frame: k["Active SKUs"] = int(frame[product_col].nunique())
    if category_col in frame: k["Active Categories"] = int(frame[category_col].nunique())
    if rating_col in frame: k["Avg Rating"] = safe_mean(frame[rating_col])
    if order_id_col in frame: k["Orders"] = int(frame[order_id_col].nunique())
    if customer_col in frame: k["Active Customers"] = int(frame[customer_col].nunique())
    # Order economics
    if order_id_col in frame and value_col in frame:
        order_sales = frame.groupby(order_id_col)[value_col].sum()
        k["AOV (Avg Order Value)"] = safe_mean(order_sales)
    return k

def kpi_row(frame, label_col, label_val):
    out = {"Label": label_val}
    out.update(compute_kpis(frame))
    return out

def format_kpis(df_kpi: pd.DataFrame) -> pd.DataFrame:
    pct_cols = [c for c in df_kpi.columns if c.endswith("%")]
    for c in pct_cols:
        df_kpi[c] = df_kpi[c].map(lambda x: f"{x:.2f}%" if pd.notna(x) else "")
    num_cols = [c for c in df_kpi.columns if c not in ["Label"] + pct_cols]
    for c in num_cols:
        if c in df_kpi:
            df_kpi[c] = df_kpi[c].map(lambda x: f"{x:,.2f}" if pd.notna(x) else "")
    return df_kpi



```

```python
# =========================================================
# Generic Descriptive Analytics Toolkit (Weekly, Flexible)
# =========================================================
# Works across varied ecommerce/retail datasets by:
# - schema inference with synonyms & date detection
# - safe numeric conversions
# - modular KPIs
# ---------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from io import StringIO
from google.colab import files
pd.set_option("display.max_columns", 120)

# -----------------------------
# 0) Upload CSV
# -----------------------------
print("Please upload your CSV file...")
uploaded = files.upload()
filename = next(iter(uploaded))
df = pd.read_csv(StringIO(uploaded[filename].decode("utf-8", errors="ignore")))
print(f"File '{filename}' loaded. Shape: {df.shape}")
display(df.head())

# -----------------------------
# 1) CONFIG (override here if needed)
# -----------------------------
CONFIG = {
    # Week anchor: "W-MON" or "W-SUN"
    "WEEK_FREQ": "W-MON",
    # On-time SLA in days for fulfillment KPIs
    "SLA_DAYS": 5,
    # % threshold to flag WoW change in action notes
    "WOW_ALERT_THRESH": 5.0,
    # Max categories to show in sheets/tables
    "TOP_CATEGORIES": 25,
}

# If the auto-mapper misses your columns, you can force them here:
OVERRIDE = {
    # Example: "price": "unit_price", "quantity": "qty", "date": "created_at"
    # Leave empty to use inference.
}
for k, v in OVERRIDE.items():
    mapping[k] = v

# -----------------------------
# 2) Schema inference utilities
# -----------------------------
def find_col(cols, candidates):
    """Return first column that matches any candidate (substring match, case-insensitive)."""
    cols_l = [c.lower() for c in cols]
    for cand in candidates:
        cand_l = cand.lower()
        for i, c in enumerate(cols_l):
            if cand_l in c:
                return cols[i]
    return None

def guess_datetime_col(frame):
    """Try to find a datetime-like column by attempting to parse each column; pick the one with most parses."""
    best_col, best_valid = None, 0
    for c in frame.columns:
        s = pd.to_datetime(frame[c], errors="coerce", infer_datetime_format=True)
        valid = s.notna().sum()
        if valid > best_valid and valid > 0:
            best_col, best_valid = c, valid
    return best_col

cols = df.columns.tolist()

# Synonym dictionary (broad)
SYN = {
    "date": ["order_date","date","purchase","created_at","timestamp","order_ts","datetime","time"],
    "delivery_date": ["delivery_date","delivered_at","ship_date","shipping_date","fulfillment_date"],
    "price": ["discounted_price","selling_price","unit_price","price","sale_price","net_price"],
    "actual_price": ["actual_price","original_price","mrp","list_price","base_price","regular_price"],
    "quantity": ["quantity","qty","units","order_item_quantity","unit_sold","units_sold"],
    "revenue": ["revenue","sales","amount","order_value","gross_sales","net_sales","gmv"],
    "cost": ["cost","cogs","cost_of_goods_sold"],
    "profit": ["profit","margin_amount","gross_profit","net_profit"],
    "discount_value": ["discount","discount_amount","discount_value","markdown"],
    "refund": ["refund","returns","refund_amount","return_amount","cancellation_amount"],
    "category": ["category","product_category","main_category","dept","department"],
    "product": ["product","product_id","asin","sku","title","item_name","product_title","item"],
    "channel": ["channel","marketplace","platform","source","utm_source","store","region"],
    "rating": ["rating","review_rating","stars","product_rating","avg_rating"],
    "order_id": ["order_id","order","txn_id","transaction_id","order_number"],
    "customer_id": ["customer_id","user_id","buyer_id","client_id","member_id"]
}

# Build initial mapping
mapping = {}
for key, cand in SYN.items():
    mapping[key] = find_col(cols, cand)

# Date fallback: guess if needed
if mapping["date"] is None:
    mapping["date"] = guess_datetime_col(df)

# Apply manual overrides if provided
for k, v in OVERRIDE.items():
    mapping[k] = v

print("Inferred/Overridden column mapping:")
print(mapping)

# -----------------------------
# 3) Preprocess: numeric coercion & time
# -----------------------------
def to_num(s):
    return pd.to_numeric(s, errors="coerce")

# Core value logic: prefer revenue; else price*qty; else price
value_col = None
if mapping["revenue"]:
    value_col = mapping["revenue"]
elif mapping["price"] and mapping["quantity"]:
    df["_value"] = to_num(df[mapping["price"]]) * to_num(df[mapping["quantity"]])
    value_col = "_value"
else:
    value_col = mapping["price"]  # fallback (some KPIs still work)

# Parse dates + weekly bins
if mapping["date"]:
    df[mapping["date"]] = pd.to_datetime(df[mapping["date"]], errors="coerce", infer_datetime_format=True)
    wk = df[mapping["date"]].dt.to_period(CONFIG["WEEK_FREQ"])
    df["_week"] = wk.astype(str)
    # elementwise start_time for pandas 2.x compatibility
    df["_week_start"] = wk.apply(lambda p: p.start_time.date() if pd.notnull(p) else np.nan)
else:
    df["_week"] = "All"
    df["_week_start"] = "All"

# -----------------------------
# 4) KPI functions (generic)
# -----------------------------
def safe_sum(series): return float(to_num(series).sum()) if series is not None else np.nan
def safe_mean(series):
    s = to_num(series)
    return float(s.mean()) if s.notna().any() else np.nan

def compute_kpis(frame: pd.DataFrame) -> dict:
    k = {}
    # Sales & units
    if value_col in frame:
        k["Total Sales"] = safe_sum(frame[value_col])
        k["Avg Sales/Row"] = safe_mean(frame[value_col])
    if mapping["quantity"] in frame:
        k["Total Units"] = safe_sum(frame[mapping["quantity"]])
        if mapping["order_id"] in frame:
            n_orders = frame[mapping["order_id"]].nunique()
            k["UPT (Units/Txn)"] = (k["Total Units"]/n_orders) if n_orders else np.nan
    # Price & margins
    if mapping["price"] in frame:
        k["ASP (Avg Selling Price)"] = safe_mean(frame[mapping["price"]])
    if mapping["profit"] in frame:
        k["Total Profit"] = safe_sum(frame[mapping["profit"]])
    if mapping["profit"] in frame and value_col in frame:
        sales = k.get("Total Sales", safe_sum(frame[value_col]))
        profit = safe_sum(frame[mapping["profit"]])
        k["Gross Margin %"] = (profit / sales * 100) if sales else np.nan
    # Discounts
    if mapping["actual_price"] and mapping["price"] in frame:
        ap = to_num(frame[mapping["actual_price"]])
        sp = to_num(frame[mapping["price"]])
        pct_off = (ap - sp) / ap
        k["Average Discount %"] = float(pct_off.replace([np.inf, -np.inf], np.nan).mean() * 100) if pct_off.notna().any() else np.nan
        k["Total Discount"] = float((ap - sp).clip(lower=0).sum())
        if value_col in frame:
            total_sales = safe_sum(frame[value_col])
            k["Discount % of Sales"] = (k["Total Discount"]/total_sales*100) if total_sales else np.nan
        # Margin proxy if cost not available
        retained = sp / ap
        k["Margin Proxy %"] = float(retained.replace([np.inf, -np.inf], np.nan).mean()*100) if retained.notna().any() else np.nan
    elif mapping["discount_value"] in frame and value_col in frame:
        total_disc = safe_sum(frame[mapping["discount_value"]])
        total_sales = safe_sum(frame[value_col])
        k["Total Discount"] = total_disc
        k["Discount % of Sales"] = (total_disc / total_sales * 100) if total_sales else np.nan
    # Refunds
    if mapping["refund"] in frame and value_col in frame:
        refunds = safe_sum(frame[mapping["refund"]])
        sales = safe_sum(frame[value_col])
        k["Refunds"] = refunds
        k["Refund Rate %"] = (refunds/sales*100) if sales else np.nan
    # Breadth & customers
    if mapping["product"] in frame: k["Active SKUs"] = int(frame[mapping["product"]].nunique())
    if mapping["category"] in frame: k["Active Categories"] = int(frame[mapping["category"]].nunique())
    if mapping["rating"] in frame: k["Avg Rating"] = safe_mean(frame[mapping["rating"]])
    if mapping["order_id"] in frame: k["Orders"] = int(frame[mapping["order_id"]].nunique())
    if mapping["customer_id"] in frame: k["Active Customers"] = int(frame[mapping["customer_id"]].nunique())
    # AOV
    if mapping["order_id"] in frame and value_col in frame:
        tmp = frame[[mapping["order_id"], value_col]].copy()
        tmp[value_col] = to_num(tmp[value_col])
        order_sales = tmp.groupby(mapping["order_id"])[value_col].sum()
        k["AOV (Avg Order Value)"] = float(order_sales.mean()) if len(order_sales) else np.nan
    # Channel share
    if mapping["channel"] in frame and value_col in frame:
        ch = frame.groupby(mapping["channel"])[value_col].sum().sort_values(ascending=False)
        total = ch.sum()
        if total:
            k["Top Channel"] = str(ch.index[0])
            k["Top Channel Share %"] = float(ch.iloc[0] / total * 100)
    return k

def kpi_row(frame, label):
    row = {"Label": label}
    row.update(compute_kpis(frame))
    return row

def format_kpis(table: pd.DataFrame) -> pd.DataFrame:
    tbl = table.copy()
    pct_cols = [c for c in tbl.columns if c.endswith("%")]
    for c in pct_cols:
        tbl[c] = tbl[c].map(lambda x: f"{x:.2f}%" if pd.notna(x) else "")
    num_cols = [c for c in tbl.columns if c not in ["Label"] + pct_cols]
    for c in num_cols:
        if c in tbl:
            tbl[c] = tbl[c].map(lambda x: f"{x:,.2f}" if pd.notna(x) else "")
    return tbl

# -----------------------------
# 5) Overall & Weekly KPI tables
# -----------------------------
overall_kpis = pd.DataFrame([kpi_row(df, "Overall")])
display(format_kpis(overall_kpis.copy()))

weekly_rows = [kpi_row(g, wk) for wk, g in df.groupby("_week_start")]
weekly_kpis = pd.DataFrame(weekly_rows).sort_values("Label")

cols_show = [c for c in [
    "Label","Total Sales","Orders","AOV (Avg Order Value)","Total Units","UPT (Units/Txn)",
    "ASP (Avg Selling Price)","Total Profit","Gross Margin %",
    "Total Discount","Average Discount %","Discount % of Sales","Margin Proxy %",
    "Refunds","Refund Rate %",
    "Active SKUs","Active Categories","Avg Rating","Active Customers",
    "Top Channel","Top Channel Share %"
] if c in weekly_kpis.columns]
display(format_kpis(weekly_kpis[cols_show].copy()).head(25))

# -----------------------------
# 6) Trend (WoW) on primary metric
# -----------------------------
metric = "Total Sales" if "Total Sales" in weekly_kpis.columns else (
    next((c for c in weekly_kpis.columns if c not in ["Label"]), None)
)

trend = weekly_kpis[["Label", metric]].copy().rename(columns={"Label": "WeekStart"})
trend["Prev"] = trend[metric].shift(1)
trend["Abs Change"] = trend[metric] - trend["Prev"]
trend["Percent Change"] = np.where(trend["Prev"].notna(),
                                   (trend[metric] - trend["Prev"]) / trend["Prev"] * 100,
                                   np.nan)
display(trend.head(15))

plt.figure()
plt.plot(trend["WeekStart"], trend[metric], marker="o")
plt.title(f"{metric} by Week")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# -----------------------------
# 7) Benchmarking & Actionable Notes
# -----------------------------
baseline = trend[metric].median()
target = baseline * 1.05
trend["Target"] = target
trend["On Track?"] = np.where(trend[metric] >= target, "Yes", "No")
display(trend.tail(15))

notes = []
thr = CONFIG["WOW_ALERT_THRESH"]
for _, r in trend.iterrows():
    msg = []
    if pd.notna(r["Percent Change"]):
        if r["Percent Change"] < -thr: msg.append(f"{metric} down more than {thr}% WoW; investigate.")
        elif r["Percent Change"] >  thr: msg.append(f"{metric} up more than {thr}% WoW; reinforce drivers.")
    notes.append({"WeekStart": r["WeekStart"], "Notes": " | ".join(msg) if msg else "No major changes."})
action_notes = pd.DataFrame(notes)
display(action_notes.head(15))

# -----------------------------
# 8) Stakeholder Views (Channel & Category)
# -----------------------------
if mapping["channel"] and mapping["channel"] in df.columns:
    ch_rows = [kpi_row(g, ch) for ch, g in df.groupby(mapping["channel"])]
    if ch_rows:
        ch_kpis = pd.DataFrame(ch_rows).sort_values("Label")
        display(format_kpis(ch_kpis.copy()))
        if value_col and value_col in df:
            ch_share = df.groupby(mapping["channel"])[value_col].sum().sort_values(ascending=False)
            if ch_share.sum() > 0:
                 ch_share_pct = (ch_share / ch_share.sum() * 100).round(2)
                 display(pd.DataFrame({"Channel": ch_share_pct.index, "Sales Share %": ch_share_pct.values}))
    else:
        print("No data to group by channel.")
else:
    print("Channel column not inferred for analysis.")

if mapping["category"] and mapping["category"] in df.columns:
    cat_rows = [kpi_row(g, c) for c, g in df.groupby(mapping["category"])]
    if cat_rows:
        cat_kpis = pd.DataFrame(cat_rows)
        if "Total Sales" in cat_kpis:
            cat_kpis = cat_kpis.sort_values("Total Sales", ascending=False).head(CONFIG["TOP_CATEGORIES"])
        display(format_kpis(cat_kpis.copy()))
    else:
        print("No data to group by category.")
else:
    print("Category column not inferred for analysis.")


# -----------------------------
# 9) Advanced KPIs (generic)
# -----------------------------
print("\n--- Advanced KPIs ---")

# 9.1 WASP
wasp = np.nan
if mapping["price"] and mapping["price"] in df.columns and (mapping["quantity"] in df.columns):
    price_num = to_num(df[mapping["price"]])
    qty_num = to_num(df[mapping["quantity"]])
    total_units = qty_num.sum()
    if total_units > 0:
        wasp = (price_num * qty_num).sum() / total_units
print("Weighted Avg Selling Price (WASP):", wasp)

# 9.2 Elasticity proxy (Discount% vs Units corr)
elasticity_corr = np.nan
if mapping["actual_price"] and mapping["actual_price"] in df.columns and mapping["price"] and mapping["price"] in df.columns and (mapping["quantity"] in df.columns):
    ap = to_num(df[mapping["actual_price"]])
    sp = to_num(df[mapping["price"]])
    qty_num = to_num(df[mapping["quantity"]])
    # Handle division by zero/inf/nan explicitly
    disc_pct = (ap - sp) / ap
    disc_pct = disc_pct.replace([np.inf, -np.inf], np.nan)

    valid_data = pd.concat([disc_pct.rename("disc_pct"), qty_num.rename("units")], axis=1).dropna()

    if len(valid_data) > 1:
        # Calculate correlation only on valid data
        elasticity_corr = valid_data["disc_pct"].corr(valid_data["units"])
print("Discount % vs Units correlation (elasticity proxy):", elasticity_corr)


# 9.3 Rating KPIs
rating_summary = {}
if mapping["rating"] and mapping["rating"] in df.columns:
    r = to_num(df[mapping["rating"]])
    # Handle cases where rating is all NaN
    if r.notna().sum() > 0:
      dist = r.dropna().round().value_counts(normalize=True).sort_index() * 100
      rating_summary["Distribution % (by star)"] = dist

      if value_col and value_col in df:
          value_num = to_num(df[value_col])
          if value_num.notna().sum() > 0:
             wr = (r * value_num).sum() / value_num.sum()
             rating_summary["Weighted Avg Rating (by sales)"] = wr
      elif mapping["quantity"] and mapping["quantity"] in df.columns:
          qty_num = to_num(df[mapping["quantity"]])
          if qty_num.notna().sum() > 0:
             wr = (r * qty_num).sum() / qty_num.sum()
             rating_summary["Weighted Avg Rating (by units)"] = wr

      if mapping["product"] and mapping["product"] in df.columns:
          # Calculate average rating per product, dropping products with no rating data
          prod_avg = df.groupby(mapping["product"])[mapping["rating"]].mean(numeric_only=True).dropna()
          if len(prod_avg) > 0:
            rating_summary["Products <3.5 Stars (%)"] = (prod_avg < 3.5).mean() * 100
      print("Rating KPIs:")
      for k, v in rating_summary.items():
          if isinstance(v, pd.Series):
              display(v.rename_axis("Stars").reset_index(name="Share %"))
          else:
              print(k, ":", v)
    else:
        print("Rating column found but contains no valid data.")

else:
    print("No rating column inferred or found.")


# 9.4 Fulfillment KPIs
fulfillment_summary = {}
if mapping["date"] and mapping["date"] in df.columns and mapping["delivery_date"] and mapping["delivery_date"] in df.columns:
    # Extract Series directly and create a new DataFrame with unique column names
    # Ensure original columns are strings before conversion attempt
    order_dates_str = df[mapping["date"]].astype(str)
    delivery_dates_str = df[mapping["delivery_date"]].astype(str)

    # Convert to datetime Series, coercing errors
    order_dates_dt = pd.to_datetime(order_dates_str, errors="coerce", infer_datetime_format=True)
    delivery_dates_dt = pd.to_datetime(delivery_dates_str, errors="coerce", infer_datetime_format=True)

    # Create a new DataFrame from the converted Series, dropna on the new datetime columns
    tmp = pd.DataFrame({
        "order_date_temp": order_dates_dt,
        "delivery_date_temp": delivery_dates_dt
    }).dropna().copy()

    if len(tmp) > 0:
        tmp["lead_days"] = (tmp["delivery_date_temp"] - tmp["order_date_temp"]).dt.days

        # Filter out negative lead times which are likely data errors
        tmp = tmp[tmp["lead_days"] >= 0].copy()

        if len(tmp) > 0:
            fulfillment_summary["Avg Delivery Time (days)"] = tmp["lead_days"].mean()
            fulfillment_summary["Delivery Time Std (days)"] = tmp["lead_days"].std()
            sla = CONFIG["SLA_DAYS"]
            fulfillment_summary[f"On-time Rate ≤ {sla} days (%)"] = (tmp["lead_days"] <= sla).mean() * 100
            fulfillment_summary["Late Delivery Rate (%)"] = (tmp["lead_days"] > sla).mean() * 100
            print("Fulfillment KPIs:")
            for k, v in fulfillment_summary.items():
                    print(k, ":", v)
        else:
            print("Fulfillment KPIs skipped: No valid positive lead times found after parsing.")
    else:
        print("Fulfillment KPIs skipped: Date conversion failed for all rows or no rows with both dates found.")
else:
    print("Fulfillment KPIs skipped (need date and delivery_date columns).")


# 9.5 Long-tail share (SKU concentration)
long_tail_share = np.nan
if mapping["product"] and mapping["product"] in df.columns and value_col and value_col in df:
    # Ensure value_col is numeric
    df[value_col] = to_num(df[value_col])
    # Drop NA for relevant columns before grouping
    valid_df_for_longtail = df.dropna(subset=[mapping["product"], value_col])
    if len(valid_df_for_longtail) > 0:
        sku_sales = valid_df_for_longtail.groupby(mapping["product"])[value_col].sum().sort_values(ascending=False)
        if sku_sales.sum() > 0:
            cum_share = sku_sales.cumsum() / sku_sales.sum()
            # Find the point where cumulative share exceeds 80%
            top_80_index = cum_share[cum_share <= 0.8].index
            if not top_80_index.empty:
                # Calculate share of items *after* the 80% cutoff
                long_tail_share = 1 - cum_share[top_80_index[-1]]
            else:
                # If no SKU gets to 80%, the long tail is the entire dataset
                long_tail_share = 1.0
        else:
            long_tail_share = 0.0 # No sales
    else:
        long_tail_share = np.nan # No valid data for calculation

print("Long-tail Share (%):", long_tail_share * 100 if pd.notna(long_tail_share) else np.nan)


# 9.6 Category Performance Index
cat_perf_index = pd.DataFrame()
if mapping["category"] and mapping["category"] in df.columns and value_col and value_col in df:
    # Ensure value_col is numeric
    df[value_col] = to_num(df[value_col])
    # Drop NA for relevant columns before grouping
    valid_df_for_cat_perf = df.dropna(subset=[mapping["category"], value_col])
    if len(valid_df_for_cat_perf) > 0:
        cat_sales = valid_df_for_cat_perf.groupby(mapping["category"])[value_col].sum()
        if cat_sales.mean() > 0: # Avoid division by zero if mean is 0
           cat_perf_index = (cat_sales / cat_sales.mean() * 100).sort_values(ascending=False).rename("Category Performance Index").reset_index()
           display(cat_perf_index.head(20))
        else:
           print("Category Performance Index skipped: Average category sales is zero.")
    else:
        print("Category Performance Index skipped: No valid data for calculation.")

else:
    print("Category Performance Index skipped (need category and value columns).")


# 9.7 Weekly MA & Volatility
weekly_ma = pd.DataFrame()
if mapping["date"] and mapping["date"] in df.columns and value_col and value_col in df:
    # Ensure date column is datetime and value column is numeric
    df[mapping["date"]] = pd.to_datetime(df[mapping["date"]], errors="coerce")
    df[value_col] = to_num(df[value_col])

    # Drop rows where date or value is missing for time series analysis
    valid_df_for_ma = df.dropna(subset=[mapping["date"], value_col]).copy()

    if len(valid_df_for_ma) > 0:
        # Group by the inferred date column after ensuring it's datetime
        wk_sales = valid_df_for_ma.set_index(mapping["date"]).groupby(pd.Grouper(freq=CONFIG["WEEK_FREQ"]))[value_col].sum().rename("sales").reset_index()

        if len(wk_sales) > 1: # Need at least 2 weeks for MA/Volatility
            wk_sales["MA_4w"] = wk_sales["sales"].rolling(window=4, min_periods=1).mean()
            wk_sales["Volatility_σ_8w"] = wk_sales["sales"].rolling(window=8, min_periods=2).std()
            weekly_ma = wk_sales.copy()
            display(weekly_ma.head(12))

            plt.figure()
            plt.plot(weekly_ma[mapping["date"]], weekly_ma["sales"], marker="o", label="Sales")
            plt.plot(weekly_ma[mapping["date"]], weekly_ma["MA_4w"], label="4w MA")
            plt.xticks(rotation=45, ha="right")
            plt.legend()
            plt.title("Weekly Sales and 4-week Moving Average")
            plt.tight_layout()
            plt.show()
        else:
            print("Weekly MA & Volatility skipped: Need at least 2 weeks of data with sales.")
    else:
         print("Weekly MA & Volatility skipped: No valid date and sales data found.")

else:
    print("Weekly MA & Volatility skipped (need date and value columns).")


# 9.8 Category Mix Shift (WoW)
mix_change = pd.DataFrame()
if mapping["date"] and mapping["date"] in df.columns and mapping["category"] and mapping["category"] in df.columns and value_col and value_col in df:
    # Ensure date and value columns are correct types
    df[mapping["date"]] = pd.to_datetime(df[mapping["date"]], errors="coerce")
    df[value_col] = to_num(df[value_col])

    # Drop NA for relevant columns
    valid_df_for_mix = df.dropna(subset=[mapping["date"], mapping["category"], value_col]).copy()

    if len(valid_df_for_mix) > 0:
        # Group by week and category
        wkcat = valid_df_for_mix.set_index(mapping["date"]).groupby([pd.Grouper(freq=CONFIG["WEEK_FREQ"]), mapping["category"]])[value_col].sum().reset_index()

        # Calculate total sales by week
        total_by_week = wkcat.groupby(mapping["date"])[value_col].transform("sum")

        # Avoid division by zero if a week has zero total sales
        wkcat["Share%"] = np.where(total_by_week > 0, wkcat[value_col] / total_by_week * 100, 0)

        # Pivot to get categories as columns and weeks as index
        mix = wkcat.pivot(index=mapping["date"], columns=mapping["category"], values="Share%").sort_index()

        # Calculate week-over-week change
        mix_change = mix.diff(periods=1)
        display(mix_change.head(10))
    else:
        print("Category Mix Shift skipped: No valid data found for calculation.")

else:
    print("Category Mix Shift skipped (need date, category, and value columns).")


# 9.9 Correlations
corr_sales_rating = np.nan
corr_price_rating = np.nan
if mapping["rating"] and mapping["rating"] in df.columns:
    r = to_num(df[mapping["rating"]])
    if r.notna().sum() > 1: # Need at least 2 non-NaN values for correlation
        if value_col and value_col in df:
            value_num = to_num(df[value_col])
            valid_corr_s = pd.concat([r.rename("rating"), value_num.rename("sales")], axis=1).dropna()
            if len(valid_corr_s) > 1:
                 corr_sales_rating = valid_corr_s.corr().iloc[0,1]
            else:
                 print("Correlation (Rating vs Sales) skipped: Not enough overlapping non-missing data.")

        if mapping["price"] and mapping["price"] in df.columns:
            price_num = to_num(df[mapping["price"]])
            valid_corr_p = pd.concat([r.rename("rating"), price_num.rename("price")], axis=1).dropna()
            if len(valid_corr_p) > 1:
                corr_price_rating = valid_corr_p.corr().iloc[0,1]
            else:
                print("Correlation (Rating vs Price) skipped: Not enough overlapping non-missing data.")
    else:
        print("Correlations skipped: Rating column contains less than 2 valid values.")

else:
    print("Correlations skipped (need rating column).")

print("Correlation (Rating vs Sales):", corr_sales_rating)
print("Correlation (Rating vs Price):", corr_price_rating)


# -----------------------------
# 10) Export to Excel (multi-sheet)
# -----------------------------
out = "generic_descriptive_analytics_weekly.xlsx"
with pd.ExcelWriter(out) as xw:
    overall_kpis.to_excel(xw, index=False, sheet_name="Overall_KPIs")
    if 'weekly_kpis' in locals() and len(weekly_kpis):
        weekly_kpis.to_excel(xw, index=False, sheet_name="Weekly_KPIs")
    if 'trend' in locals() and len(trend):
        trend.to_excel(xw, index=False, sheet_name="Trend_WoW")
    if 'ch_kpis' in locals() and len(ch_kpis):
        ch_kpis.to_excel(xw, index=False, sheet_name="Channel_KPIs")
    if 'cat_kpis' in locals() and len(cat_kpis):
        cat_kpis.to_excel(xw, index=False, sheet_name="Category_KPIs")

    # Advanced Summary sheet
    advanced_summary_data = {
        "WASP":[wasp],
        "Elasticity Corr":[elasticity_corr],
        "LongTail%":[long_tail_share*100 if pd.notna(long_tail_share) else np.nan],
        "Corr_Rating_Sales":[corr_sales_rating],
        "Corr_Rating_Price":[corr_price_rating],
    }
    pd.DataFrame(advanced_summary_data).to_excel(xw, index=False, sheet_name="Advanced_Summary")

    if isinstance(rating_summary.get("Distribution % (by star)"), pd.Series):
        rating_summary["Distribution % (by star)"].rename_axis("Stars").reset_index(name="Share %").to_excel(xw, index=False, sheet_name="Rating_Distribution")
    if 'cat_perf_index' in locals() and len(cat_perf_index):
        cat_perf_index.to_excel(xw, index=False, sheet_name="Category_Perf_Index")
    if 'weekly_ma' in locals() and len(weekly_ma):
        # Ensure the date column used as index for plot is also in the Excel
        weekly_ma.to_excel(xw, index=False, sheet_name="Weekly_MA_Volatility")
    if 'mix_change' in locals() and len(mix_change):
         # Reset index to include the date column in Excel
        mix_change.reset_index().to_excel(xw, index=False, sheet_name="Category_Mix_Shift")
    if len(fulfillment_summary):
        pd.DataFrame(list(fulfillment_summary.items()), columns=["Metric","Value"]).to_excel(xw, index=False, sheet_name="Fulfillment")


files.download(out)
```

# Lesson 14. Statistical Analysis for Diagnostic Analytics

Statistical Methods:
1. Hypothesis Testing
2. Correlation Analysis
3. Regression Analysis
4. ANOVA
5. Factor Analysis
6. Principal Component Analysis (PCA)
7. Multivariate Analysis

--> Applied in combinations based on research questions

Dataset: Lesson14_healthcare_noshows.csv

Research Question:
What patient demographics, health conditions, appointment scheduling factors, and reminders influence the probability of patients missing their scheduled appointments?

Hypothesis:
H0: None of the studied factors significantly affect no-show rates
Ha: One or more factors significantly influence no-show likelihood

Method:
1. Root Cause and Association Analysis
- Descriptive statistics and visualizations for no-show rates across groups
- Chi-square and Mann-Whitney U tests to associate categorical and numeric variables with no-show
2. Multivariate Explanatory Modeling
- Logistic regression modeling to identify key predictors of no-show
- Random Forest feature importance: which variables, individually or in combination, have the strongest diagnostic impact on the outcome
- Interaction analysis: if combinations of patient and appointment features jointly contribute to no-shows

< Example > Factors Influencing Healthcare Service Time

1) Research Question:
Do patient group, medical department, and delay reason significantly affect the service time duration in healthcare appointments?

2) Hypothesis:
Null hypothesis: There are no significant differences in mean service time across different patient groups, medical departments, or delay reasons.  
Alternative hypothesis: At least one patient group, medical department, or delay reason has a significantly different mean service time.

3) Statistical Method - ANOVA
- One-Way ANOVA for each factor (Patient_Group, Department, Delay_Reason) to assess their individual effects on service time.  
- Tukey's Honestly Significant Difference (HSD) post hoc tests for pairwise group comparisons where ANOVA shows significance.  

4) Statistical Method - Kruskal-Wallis Test
- Performs Kruskal-Wallis test on service time grouped by each factor.
- If p < 0.05, runs Dunn's post hoc test (adjusted for multiple comparisons).
- Displays pairwise p-values matrix and boxplots for visual interpretation.

```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import statsmodels.stats.multicomp as mc
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
data = pd.read_csv(file_name)

for col in ['Patient_Group', 'Department', 'Delay_Reason']:
    data[col] = data[col].astype('category')

def print_anova(anova_results):
    display(anova_results.style.format({
        'sum_sq': '{:.2f}',
        'df': '{:.0f}',
        'F': '{:.4f}',
        'PR(>F)': '{:.4f}'
    }).set_caption('ANOVA Table'))

def print_posthoc(posthoc):
    df = pd.DataFrame(data=posthoc._results_table.data[1:], columns=posthoc._results_table.data[0])
    display(df.style.set_table_styles([
        {'selector': 'th', 'props': [('background-color', '#f7f7f9'), ('color', '#333')]},
        {'selector': 'td', 'props': [('padding', '6px')]}
    ]).format({
        'meandiff': '{:.3f}',
        'p-adj': '{:.4f}',
        'lower': '{:.3f}',
        'upper': '{:.3f}'
    }).set_caption('Tukey HSD Post Hoc Test Results'))

def analyze_factor(factor):
    print(f"\nAnalyzing factor: {factor}")
    model = smf.ols(f'Q("Service_Time(mins)") ~ C({factor})', data=data).fit()
    anova_results = sm.stats.anova_lm(model, typ=2)
    print_anova(anova_results)

    p_val = anova_results['PR(>F)'].iloc[0]
    if pd.notna(p_val) and p_val < 0.05:
        comp = mc.MultiComparison(data['Service_Time(mins)'], data[factor])
        posthoc = comp.tukeyhsd()
        print_posthoc(posthoc)

        plt.figure(figsize=(8,5))
        sns.boxplot(x=factor, y='Service_Time(mins)', data=data)
        plt.title(f'Service_Time by {factor}')
        plt.show()
    else:
        print("No significant differences found; skipping post hoc and plot.")

# Analyze each factor one-way ANOVA
for factor in ['Patient_Group', 'Department', 'Delay_Reason']:
    analyze_factor(factor)

# Diagnose counts for factor combinations to check sparsity
print("\nCounts of observations per factor combination:")
print(data.groupby(['Patient_Group', 'Department', 'Delay_Reason']).size())


```

< Example > Non-parametric Analysis

```python
import pandas as pd
from scipy.stats import kruskal
import scikit_posthocs as sp
import matplotlib.pyplot as plt
import seaborn as sns

# Upload dataset
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
data = pd.read_csv(file_name)

# Convert factors to category type
for col in ['Patient_Group', 'Department', 'Delay_Reason']:
    data[col] = data[col].astype('category')

def kruskal_wallis_test(factor):
    print(f"\nKruskal-Wallis test for {factor}")
    groups = [group['Service_Time(mins)'].values for name, group in data.groupby(factor)]
    stat, p = kruskal(*groups)
    print(f"H-statistic = {stat:.4f}, p-value = {p:.4f}")

    if p < 0.05:
        print("Significant differences detected, performing Dunn's post hoc test...")
        # Dunn's test with Bonferroni correction
        posthoc = sp.posthoc_dunn(data, val_col='Service_Time(mins)', group_col=factor, p_adjust='bonferroni')
        display(posthoc)

        # Plot boxplot
        plt.figure(figsize=(8,5))
        sns.boxplot(x=factor, y='Service_Time(mins)', data=data)
        plt.title(f'Service Time by {factor}')
        plt.show()
    else:
        print("No significant differences detected.")

# Apply Kruskal-Wallis test for each factor
for factor in ['Patient_Group', 'Department', 'Delay_Reason']:
    kruskal_wallis_test(factor)
```

```python
import itertools
import networkx as nx
import matplotlib.pyplot as plt

# ---------------------------
# Step 1: Define the graph
# ---------------------------
nodes = ["A", "B", "C", "D", "E"]

edges = [
    ("A", "B", 4),
    ("A", "C", 3),
    ("A", "D", 5),
    ("A", "E", 4),
    ("B", "C", 2),
    ("B", "D", 6),
    ("B", "E", 2),
    ("C", "D", 3),
    ("C", "E", 3),
    ("D", "E", 5)
]

# Create a graph
G = nx.Graph()
G.add_weighted_edges_from(edges)

# ---------------------------
# Step 2: Visualize the graph
# ---------------------------
plt.figure(figsize=(7, 7))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color="lightblue", node_size=1200, font_size=14)
edge_labels = nx.get_edge_attributes(G, "weight")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color="red")
plt.title("Traveling Salesman Problem Graph")
plt.show()

# ---------------------------
# Step 3: Define the cost function
# ---------------------------
def route_cost(route, graph):
    cost = 0
    for i in range(len(route) - 1):
        if graph.has_edge(route[i], route[i + 1]):
            cost += graph[route[i]][route[i + 1]]["weight"]
        else:
            return float("inf")  # Invalid route
    return cost

# ---------------------------
# Step 4: Brute Force TSP Solver
# ---------------------------
start = "A"                          # Fix the starting city
min_cost = float("inf")              # Initialize minimum cost as infinite
best_route = []                      # Store the optimal path

# Generate all possible routes starting and ending at A
for perm in itertools.permutations([n for n in nodes if n != start]):
    route = [start] + list(perm) + [start]   # Complete loop (back to start)
    cost = route_cost(route, G)              # Calculate route cost

    if cost < min_cost:                      # Check if route is cheaper
        min_cost = cost                      # Update best cost
        best_route = route                   # Save best route

print("Optimal Route: ", ' -> '.join(best_route))
print("Minimum Travel Cost: ", min_cost)

# ---------------------------
# Step 5: Visualize the optimal route
# ---------------------------
best_edges = [(best_route[i], best_route[i + 1]) for i in range(len(best_route) - 1)]

plt.figure(figsize=(7, 7))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color="lightgrey", node_size=1200)
nx.draw_networkx_edges(G, pos, edgelist=best_edges, width=3, edge_color="blue")
nx.draw_networkx_labels(G, pos, font_size=14)
edge_labels = nx.get_edge_attributes(G, "weight")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color="red")
plt.title(f"Optimal Route: {' -> '.join(best_route)}\nTotal Cost: {min_cost}")
plt.show()
```

Step 1. Define the Decision Problem


```python
import numpy as np
import pandas as pd
from pulp import *
from ortools.sat.python import cp_model
from geneticalgorithm import geneticalgorithm as ga

# Problem Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
drivers = ["Alice", "Bob", "Charlie", "Diana", "Ethan"]

driver_cost = {"Alice": 50, "Bob": 45, "Charlie": 40, "Diana": 55, "Ethan": 60}
max_shifts = {"Alice": 5, "Bob": 5, "Charlie": 4, "Diana": 3, "Ethan": 4}
demand = {"Mon": 2, "Tue": 2, "Wed": 3, "Thu": 3, "Fri": 4, "Sat": 4, "Sun": 3}

def print_schedule(x_dict, solver=None):  #  Added 'solver' parameter
    """Universal print function for both PuLP and OR-Tools results."""
    schedule = pd.DataFrame(index=days, columns=drivers)
    for d in days:
        for p in drivers:
            if solver is not None:  #  Handle CP-SAT (OR-Tools)
                val = solver.Value(x_dict[(d, p)])
            elif hasattr(x_dict[(d, p)], "varValue"):  #  Handle PuLP
                val = x_dict[(d, p)].varValue
            else:  #  Handle pre-extracted values
                val = x_dict[(d, p)]
            schedule.loc[d, p] = int(round(val))
    display(schedule)

```

Step 2. Model Formulation (Mathematical Optimization Model)


Minimize Total Cost = Σ (Cost × Assigned Shifts)

Subject to:

Minimum drivers per day (demand)

Daily and weekly driver limits


Step 3. Choose and Apply Optimization Methods

(a) Linear Programming (LP) — Continuous Decision Variables

--> Allows fractional allocations



```python
model_lp = LpProblem("LP_Scheduling", LpMinimize)
x = LpVariable.dicts("shift", [(d, p) for d in days for p in drivers], 0, 1)

# Objective
model_lp += lpSum(driver_cost[p]*x[(d, p)] for d in days for p in drivers)
# Constraints
for d in days:
    model_lp += lpSum(x[(d, p)] for p in drivers) >= demand[d]
for p in drivers:
    model_lp += lpSum(x[(d, p)] for d in days) <= max_shifts[p]

model_lp.solve(PULP_CBC_CMD(msg=0))
print("LP — Total Cost =", value(model_lp.objective))
print_schedule(x)
```

(b) Integer Programming (IP) — Binary Decision Variables

--> Schedules discrete shifts: 1 = assigned, 0 = off


```python
model_ip = LpProblem("IP_Scheduling", LpMinimize)
x = LpVariable.dicts("shift", [(d, p) for d in days for p in drivers], 0, 1, LpInteger)

model_ip += lpSum(driver_cost[p]*x[(d, p)] for d in days for p in drivers)
for d in days:
    model_ip += lpSum(x[(d, p)] for p in drivers) >= demand[d]
for p in drivers:
    model_ip += lpSum(x[(d, p)] for d in days) <= max_shifts[p]

model_ip.solve(PULP_CBC_CMD(msg=0))
print("IP — Total Cost =", value(model_ip.objective))
print_schedule(x)
```

(c) Mixed-Integer Programming (MIP) — Binary + Continuous

--> Adds continuous overtime variable for flexibility (e.g., up to 2 extra shifts)

```python
model_mip = LpProblem("MIP_Scheduling", LpMinimize)
x = LpVariable.dicts("shift", [(d, p) for d in days for p in drivers], 0, 1, LpInteger)
overtime = LpVariable.dicts("overtime", drivers, lowBound=0)

model_mip += (lpSum(driver_cost[p]*x[(d, p)] for d in days for p in drivers)
              + 20*lpSum(overtime[p] for p in drivers))
for d in days:
    model_mip += lpSum(x[(d, p)] for p in drivers) >= demand[d]
for p in drivers:
    model_mip += lpSum(x[(d, p)] for d in days) <= max_shifts[p] + overtime[p]

model_mip.solve(PULP_CBC_CMD(msg=0))
print("MIP — Total Cost =", value(model_mip.objective))
print("Overtime per Driver:")
for p in drivers:
    print(f"  {p}: {overtime[p].varValue:.2f}")
print_schedule(x)
```

(d) Constraint Optimization (CP-SAT via OR-Tools)

--> Ensures logical feasibility instead of just numeric optimization.


```python
model_cp = cp_model.CpModel()
x_cp = {}

# Define decision variables for each day and driver
for d in days:
    for p in drivers:
        x_cp[(d, p)] = model_cp.NewBoolVar(f"x_{d}_{p}")

# Define constraints
for d in days:
    model_cp.Add(sum(x_cp[(d, p)] for p in drivers) >= demand[d])
for p in drivers:
    model_cp.Add(sum(x_cp[(d, p)] for d in days) <= max_shifts[p])

# Objective: minimize total cost
model_cp.Minimize(sum(driver_cost[p] * x_cp[(d, p)] for d in days for p in drivers))

# Solve model
solver_cp = cp_model.CpSolver()
solver_cp.Solve(model_cp)

print(f"CP-SAT Total Cost: ${solver_cp.ObjectiveValue():.2f}")

# Extract results safely using solver.Value()
schedule_cp = {(d, p): solver_cp.Value(x_cp[(d, p)]) for d in days for p in drivers}
print_schedule(schedule_cp)
```

```python
import matplotlib.pyplot as plt

# ============================================================================
# 20% Increase on Friday and 30% Increase on Saturday
# ============================================================================
demand_scenario = demand.copy()  # Start with baseline demand
demand_scenario["Fri"] = int(demand["Fri"] * 1.3)  # 30% rise
demand_scenario["Sat"] = int(demand["Sat"] * 1.3)  # 30% rise
demand_scenario["Sun"] = int(demand["Sun"] * 1.4)  # 40% rise

# Print comparison summary
print("=== Scenario: Friday +20%, Saturday +30% ===\n")
print("Daily Demand Comparison:")
for d in days:
    change = demand_scenario[d] - demand[d]
    if change > 0:
        print(f"  {d}: {demand[d]} → {demand_scenario[d]} (+{change})")
    else:
        print(f"  {d}: {demand[d]} (no change)")

# ============================================================================
# Visualization: Bar Chart and Line Chart
# ============================================================================
baseline = [demand[d] for d in days]
scenario = [demand_scenario[d] for d in days]

# --- Bar Chart ---
fig, ax = plt.subplots(figsize=(9,5))
x_pos = range(len(days))
width = 0.35

ax.bar([i - width/2 for i in x_pos], baseline, width, color='skyblue', label="Baseline Demand")
ax.bar([i + width/2 for i in x_pos], scenario, width, color='tomato', label="Scenario Demand")

for i in range(len(days)):
    if baseline[i] != scenario[i]:
        ax.annotate(f"+{scenario[i]-baseline[i]}",
                    xy=(i, scenario[i]), ha='center', va='bottom', color='red', fontsize=10)

ax.set_title("Demand per Day of Week\n(Scenario: +20% Friday, +30% Saturday)",
              fontsize=13, fontweight='bold')
ax.set_xlabel("Day of Week", fontsize=11)
ax.set_ylabel("Drivers Required", fontsize=11)
ax.set_xticks(list(x_pos))
ax.set_xticklabels(days)
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# --- Line Chart ---
fig, ax = plt.subplots(figsize=(9,5))

ax.plot(days, baseline, marker='o', color='steelblue', linewidth=2, label="Baseline")
ax.plot(days, scenario, marker='s', color='tomato', linewidth=2, label="Scenario")

# Highlight differences
for i, d in enumerate(days):
    if baseline[i] != scenario[i]:
        ax.annotate('', xy=(i, scenario[i]), xytext=(i, baseline[i]),
                    arrowprops=dict(arrowstyle='->', color='red', lw=2))

ax.set_title("Daily Demand Change: Baseline → Scenario",
              fontsize=13, fontweight='bold')
ax.set_xlabel("Day of Week", fontsize=11)
ax.set_ylabel("Number of Drivers", fontsize=11)
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
demand_scenario = demand.copy()
demand_scenario["Fri"] = int(demand["Fri"] * 1.3)  # 30% rise
demand_scenario["Sat"] = int(demand["Sat"] * 1.3)  # 30% rise
demand_scenario["Sun"] = int(demand["Sun"] * 1.4)  # 40% rise

print("Scenario: Weekend Demand Spike (+20% Friday, +30% Saturday)\n")
print("Demand Changes:")
for d in days:
    baseline = demand[d]
    scenario = demand_scenario[d]
    change = scenario - baseline
    if change != 0:  #  Only print if changed
        print(f"  {d}: {baseline} → {scenario} drivers (+{change})")
    else:
        print(f"  {d}: {baseline} drivers (no change)")

#  Re-run MIP model with new demand
print("\n--- Re-optimizing with New Demand ---")
model_mip_scenario = LpProblem("MIP_Scenario", LpMinimize)
x_scenario = LpVariable.dicts("shift_sc", [(d, p) for d in days for p in drivers], 0, 1, LpInteger)
overtime_scenario = LpVariable.dicts("overtime_sc", drivers, lowBound=0, upBound=2)

model_mip_scenario += (lpSum(driver_cost[p]*x_scenario[(d, p)] for d in days for p in drivers)
                       + 20*lpSum(overtime_scenario[p] for p in drivers))

for d in days:
    model_mip_scenario += lpSum(x_scenario[(d, p)] for p in drivers) >= demand_scenario[d]  # Use new demand
for p in drivers:
    model_mip_scenario += lpSum(x_scenario[(d, p)] for d in days) <= max_shifts[p] + overtime_scenario[p]

model_mip_scenario.solve(PULP_CBC_CMD(msg=0))

#  Compare baseline vs. scenario costs
baseline_cost = value(model_mip.objective)
scenario_cost = value(model_mip_scenario.objective)
cost_increase = scenario_cost - baseline_cost

print(f"\nBaseline Total Cost: ${baseline_cost:.2f}")
print(f"Scenario Total Cost: ${scenario_cost:.2f}")
print(f"Cost Increase: ${cost_increase:.2f} (+{(cost_increase/baseline_cost)*100:.1f}%)")

#  Compare schedules
schedule_baseline = pd.DataFrame(index=days, columns=drivers)
schedule_scenario = pd.DataFrame(index=days, columns=drivers)

for d in days:
    for p in drivers:
        schedule_baseline.loc[d, p] = int(round(x[(d, p)].varValue))
        schedule_scenario.loc[d, p] = int(round(x_scenario[(d, p)].varValue))

print("\nBaseline Schedule:")
display(schedule_baseline)
print("\nScenario Schedule:")
display(schedule_scenario)

#  Show only differences
print("\nChanges in Assignments (Baseline → Scenario):")
changes_found = False
for d in days:
    for p in drivers:
        before = schedule_baseline.loc[d, p]
        after = schedule_scenario.loc[d, p]
        if before != after:
            print(f"  {d}, {p}: {before} → {after}")
            changes_found = True
if not changes_found:
    print("  No assignment changes detected.")
```

```python
import matplotlib.pyplot as plt
import pandas as pd


def plot_slope_graph_by_group_and_time(df, group_col, time_col, value_col, time1, time2, time_unit='year'):
    """
    Plot slope graph comparing average value_col by group_col between two time points (year or month).

    Args:
        df (pd.DataFrame): Dataset.
        group_col (str): Categorical column name.
        time_col (str): Datetime column name.
        value_col (str): Numeric column name.
        time1, time2 : values for comparison (int or str depending on time_unit).
        time_unit (str): 'year' or 'month' to specify aggregation.
    """
    df = df.copy()
    df[time_col] = pd.to_datetime(df[time_col], errors='coerce')
    df = df.dropna(subset=[time_col, group_col, value_col])

    if time_unit == 'year':
        df['time'] = df[time_col].dt.year
    elif time_unit == 'month':
        df['time'] = df[time_col].dt.month
    else:
        raise ValueError("time_unit must be 'year' or 'month'")

    df_filtered = df[df['time'].isin([time1, time2])]

    mean_vals = df_filtered.groupby([group_col, 'time'])[value_col].mean().reset_index()
    pivot_df = mean_vals.pivot(index=group_col, columns='time', values=value_col).reset_index()

    plt.figure(figsize=(10, 6))
    x = [time1, time2]

    for i, row in pivot_df.iterrows():
        y = [row.get(time1, float('nan')), row.get(time2, float('nan'))]
        plt.plot(x, y, marker='o', label=f'{group_col} {row[group_col]}')

    plt.xticks(x, [str(time1), str(time2)])
    plt.title(f'Slope Graph: Avg {value_col} by {group_col} ({time_unit.capitalize()} {time1} vs {time2})')
    plt.xlabel(time_unit.capitalize())
    plt.ylabel(f'Average {value_col}')
    plt.legend(title=group_col)
    plt.grid(True)
    plt.show()


def reduce_rating_categories(df):
    mapping = {
        'TV-MA': 'Adult',
        'R': 'Adult',
        'NC-17': 'Adult',
        'TV-14': 'Teen',
        'PG-13': 'Teen',
        'TV-PG': 'Teen',
        'PG': 'Kids',
        'TV-Y': 'Kids',
        'TV-Y7': 'Kids',
        'G': 'Kids',
        # Any other ratings will be grouped as 'Other'
    }

    # Use the mapping to create a reduced rating column, default to 'Other'
    df['rating_simple'] = df['rating'].map(mapping).fillna('Other')
    return df


df = reduce_rating_categories(df)

# Verify transformed rating categories
print(df['rating'].value_counts())
print(df['rating_simple'].value_counts())

# Example: Compare average number of genres by reduced rating between 2015 and 2020 release years
plot_slope_graph_by_group_and_time(df, 'rating_simple', 'date_added', 'num_genres', 2015, 2020, time_unit='year')
```

```python
import pandas as pd
import plotly.express as px


def prepare_netflix_stream_data_monthly(df, date_col='date_added', category_col='primary_genre'):
    """
    Prepare aggregated data for stream graph showing monthly counts by category.

    Args:
        df (pd.DataFrame): Netflix dataset.
        date_col (str): Column name with datetime of show added.
        category_col (str): Categorical column for grouping (e.g., genre).

    Returns:
        pd.DataFrame: Aggregated dataframe with columns [month, category_col, counts].
    """
    df = df.copy()

    # Ensure datetime format for date_col
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')
    df_clean = df.dropna(subset=[date_col, category_col]).copy()

    # Extract year-month period
    df_clean['month'] = df_clean[date_col].dt.to_period('M').dt.to_timestamp()

    # Aggregate counts by month and category
    aggregated = df_clean.groupby(['month', category_col]).size().reset_index(name='counts')

    return aggregated


def plot_stream_graph(df_agg, date_col='month', category_col='primary_genre', counts_col='counts',
                      title='Stream Graph: Counts by Category Over Time'):
    """
    Plot stream graph (stacked area chart) from aggregated data.

    Args:
        df_agg (pd.DataFrame): Aggregated data with date, category, and counts columns.
        date_col (str): Date column name.
        category_col (str): Categorical grouping column name.
        counts_col (str): Count or value column name.
        title (str): Chart title.
    """
    fig = px.area(df_agg,
                  x=date_col,
                  y=counts_col,
                  color=category_col,
                  title=title,
                  labels={date_col: 'Month', counts_col: 'Count', category_col: 'Category'})
    fig.show()

monthly_stream_data = prepare_netflix_stream_data_monthly(df, date_col='date_added', category_col='primary_genre')
plot_stream_graph(monthly_stream_data, title='Monthly Counts of Netflix Shows Added by Genre')
```

```python
import pandas as pd
import plotly.graph_objects as go
import numpy as np

def plot_sankey(df, source_col, target_col, value_col, title='Sankey Diagram'):
    agg_df = df.groupby([source_col, target_col])[value_col].sum().reset_index()

    source_labels = agg_df[source_col].unique().tolist()
    target_labels = agg_df[target_col].unique().tolist()
    all_labels = source_labels + [x for x in target_labels if x not in source_labels]

    label_to_index = {label: idx for idx, label in enumerate(all_labels)}

    sources = agg_df[source_col].map(label_to_index)
    targets = agg_df[target_col].map(label_to_index)

    values = agg_df[value_col]

    fig = go.Figure(data=[go.Sankey(
        node=dict(
            pad=15,
            thickness=20,
            line=dict(color='black', width=0.5),
            label=all_labels,
        ),
        link=dict(
            source=sources,
            target=targets,
            value=values,
            color="rgba(0,100,200,0.4)"
        )
    )])

    fig.update_layout(title_text=title, font_size=10)
    fig.show()

# Realistic sample data: payment types, taxi types, pickup boroughs, trip volumes
np.random.seed(42)

payment_types = ['Credit Card', 'Cash', 'Mobile Payment', 'Prepaid']
taxi_types = ['Yellow Taxi', 'Green Taxi', 'Livery']
pickup_boroughs = ['Manhattan', 'Brooklyn', 'Queens', 'Bronx', 'Staten Island']

# Generate synthetic flows between payment type and taxi type
data_stage1 = []
for p in payment_types:
    for t in taxi_types:
        volume = np.random.randint(100, 1000)
        data_stage1.append({'source': p, 'target': t, 'value': volume})

df_stage1 = pd.DataFrame(data_stage1)

# Generate synthetic flows between taxi type and pickup boroughs
data_stage2 = []
for t in taxi_types:
    for b in pickup_boroughs:
        volume = np.random.randint(50, 700)
        data_stage2.append({'source': t, 'target': b, 'value': volume})

df_stage2 = pd.DataFrame(data_stage2)

# Plot
plot_sankey(df_stage1, 'source', 'target', 'value', 'Payment Type to Taxi Type')
plot_sankey(df_stage2, 'source', 'target', 'value', 'Taxi Type to Pickup Borough')
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def plot_interactive_arc(df, source_col, target_col, weight_col, title='Arc Diagram (Plotly)'):
    # Extract unique nodes
    nodes = list(pd.unique(df[[source_col, target_col]].values.ravel()))
    node_indices = {node: i for i, node in enumerate(nodes)}

    # Create source, target indices
    sources = df[source_col].map(node_indices)
    targets = df[target_col].map(node_indices)
    values = df[weight_col]

    # Circular node layout
    theta = np.linspace(0, 2*np.pi, len(nodes), endpoint=False)
    x = np.cos(theta)
    y = np.sin(theta)

    # Create node scatter points
    node_trace = go.Scatter(
        x=x,
        y=y,
        mode='markers+text',
        text=nodes,
        textposition='top center',
        marker=dict(size=20, color='skyblue'),
        hoverinfo='text'
    )

    # Create arcs between nodes
    edge_traces = []

    for s, t, val in zip(sources, targets, values):
        # Compute midpoint for arc
        x0, y0 = x[s], y[s]
        x1, y1 = x[t], y[t]
        midpoint = ((x0 + x1)/2, (y0 + y1)/2)
        # Control point for Bezier curve arc (lifted above midpoint)
        xc, yc = midpoint[0]*1.2, midpoint[1]*1.2

        # Create quadratic Bezier curve points
        from scipy.special import comb

        def bezier(t, P0, P1, P2):
            return (1-t)**2 * P0 + 2*(1-t)*t * P1 + t**2 * P2

        ts = np.linspace(0,1,50)
        xs = bezier(ts, x0, xc, x1)
        ys = bezier(ts, y0, yc, y1)

        edge_trace = go.Scatter(
            x=xs,
            y=ys,
            mode='lines',
            line=dict(width=val/3, color='royalblue'),
            hoverinfo='none',
            opacity=0.7
        )
        edge_traces.append(edge_trace)

    fig = go.Figure(data=[node_trace] + edge_traces)
    fig.update_layout(title=title, showlegend=False,
                      xaxis=dict(showgrid=False, zeroline=False, visible=False),
                      yaxis=dict(showgrid=False, zeroline=False, visible=False),
                      plot_bgcolor='white', margin=dict(t=40, b=0, l=0, r=0))
    fig.show()

# Sample  social network data
np.random.seed(123)
names = ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy']
num_edges = 25
edges = []
for _ in range(num_edges):
    p1, p2 = np.random.choice(names, 2, replace=False)
    interactions = np.random.randint(1, 10)
    edges.append({'PersonA': p1, 'PersonB': p2, 'interaction_count': interactions})
df_social = pd.DataFrame(edges)

# Call the interactive arc diagram function
plot_interactive_arc(df_social, 'PersonA', 'PersonB', 'interaction_count', 'Interactive Social Network Arc Diagram')
```

Force-directed Graph

```python
import networkx as nx
import random
import plotly.graph_objects as go

def create_social_network_graph(seed=42):
    random.seed(seed)
    G = nx.Graph()

    communities = {
        'Community A': [f'Alice{i}' for i in range(1, 7)],
        'Community B': [f'Bob{i}' for i in range(1, 5)],
        'Community C': [f'Charlie{i}' for i in range(1, 6)],
    }

    for community, members in communities.items():
        for member in members:
            G.add_node(member, community=community)

    for community, members in communities.items():
        for i in range(len(members)):
            for j in range(i + 1, len(members)):
                if random.random() < 0.6:
                    G.add_edge(members[i], members[j])

    cross_community_edges = [
        ('Alice1', 'Bob1'),
        ('Alice4', 'Charlie3'),
        ('Bob3', 'Charlie5'),
    ]
    G.add_edges_from(cross_community_edges)

    return G

def plot_force_directed_graph(G):
    pos = nx.spring_layout(G, seed=42)

    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = pos[edge[0]]
        x1, y1 = pos[edge[1]]
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        line=dict(width=1, color='#888'),
        hoverinfo='none',
        mode='lines')

    node_x = []
    node_y = []
    node_color = []
    node_text = []

    community_colors = {
        'Community A': 'red',
        'Community B': 'green',
        'Community C': 'blue'
    }

    for node, data in G.nodes(data=True):
        x, y = pos[node]
        node_x.append(x)
        node_y.append(y)
        node_color.append(community_colors.get(data['community'], 'gray'))
        node_text.append(f"{node} ({data['community']})")

    node_trace = go.Scatter(
        x=node_x,
        y=node_y,
        mode='markers+text',
        text=node_text,
        textposition='top center',
        hoverinfo='text',
        marker=dict(
            showscale=False,
            color=node_color,
            size=20,
            line_width=2
        )
    )

    fig = go.Figure(data=[edge_trace, node_trace],
                    layout=go.Layout(
                        title="Social Network Force-Directed Graph",
                        title_x=0.5,
                        showlegend=False,
                        hovermode='closest',
                        margin=dict(b=20, l=5, r=5, t=40),
                        xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
                        yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
                    ))
    fig.show()

# Example usage:
G = create_social_network_graph()
plot_force_directed_graph(G)
```

```python
import networkx as nx
import plotly.graph_objects as go

def create_balanced_tree(r=3, h=3):
    """
    Create a balanced tree graph with 'subset' attribute for multipartite layout.

    Args:
        r (int): Branching factor.
        h (int): Height of tree.

    Returns:
        networkx.Graph: Balanced tree with subset attributes.
    """
    G = nx.balanced_tree(r, h)

    # Assign subset attribute based on depth
    depth_map = {node: nx.shortest_path_length(G, source=0, target=node) for node in G.nodes()}
    nx.set_node_attributes(G, depth_map, 'subset')

    return G

def get_multipartite_pos(G, subset_key='subset'):
    """
    Compute multipartite layout positions for the graph.

    Args:
        G (networkx.Graph): Input graph.
        subset_key (str): Node attribute defining subsets/layers.

    Returns:
        dict: Node positions.
    """
    pos = nx.multipartite_layout(G, subset_key=subset_key)
    return pos

def plot_radial_tree(G, pos):
    """
    Plot radial tree/network graph using Plotly.

    Args:
        G (networkx.Graph): Input graph.
        pos (dict): Node positions.
    """
    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = pos[edge[0]]
        x1, y1 = pos[edge[1]]
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        line=dict(width=2, color='#888'),
        hoverinfo='none',
        mode='lines')

    node_x = []
    node_y = []
    node_text = []
    for node in G.nodes():
        x, y = pos[node]
        node_x.append(x)
        node_y.append(y)
        node_text.append(f"Node {node}")

    node_trace = go.Scatter(
        x=node_x, y=node_y,
        mode='markers+text',
        text=node_text,
        textposition="top center",
        hoverinfo='text',
        marker=dict(
            size=10,
            color='LightSkyBlue',
            line_width=2))

    fig = go.Figure(data=[edge_trace, node_trace],
                    layout=go.Layout(
                        title="Radial Tree/Network Graph with Multipartite Layout",
                        title_x=0.5,
                        showlegend=False,
                        hovermode='closest',
                        margin=dict(b=20, l=5, r=5, t=40),
                        xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
                        yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
                    ))

    fig.show()


# Example usage:
G = create_balanced_tree(r=3, h=3)
pos = get_multipartite_pos(G)
plot_radial_tree(G, pos)


```

```python
world_edges = [
    # Level 1: Continent
    ('World', 'Africa'), ('World', 'Americas'), ('World', 'Asia'),
    ('World', 'Europe'), ('World', 'Oceania'),

    # Level 2: Regions (selected examples)
    ('Africa', 'Northern Africa'), ('Africa', 'Western Africa'), ('Africa', 'Eastern Africa'),
    ('Americas', 'North America'), ('Americas', 'Central America'), ('Americas', 'South America'),
    ('Asia', 'Western Asia'), ('Asia', 'Eastern Asia'), ('Asia', 'Southern Asia'),
    ('Europe', 'Northern Europe'), ('Europe', 'Southern Europe'), ('Europe', 'Western Europe'),
    ('Oceania', 'Australia and NZ'), ('Oceania', 'Melanesia'), ('Oceania', 'Micronesia'),

    # Level 3: Selected example countries
    ('Northern Africa', 'Egypt'), ('Northern Africa', 'Morocco'),
    ('Western Africa', 'Nigeria'), ('Western Africa', 'Ghana'),
    ('North America', 'United States'), ('North America', 'Canada'), ('Central America', 'Mexico'),
    ('South America', 'Brazil'), ('South America', 'Argentina'),
    ('Western Asia', 'Turkey'), ('Western Asia', 'Saudi Arabia'),
    ('Eastern Asia', 'China'), ('Eastern Asia', 'Japan'),
    ('Southern Asia', 'India'), ('Southern Asia', 'Pakistan'),
    ('Northern Europe', 'United Kingdom'), ('Northern Europe', 'Ireland'),
    ('Southern Europe', 'Italy'), ('Southern Europe', 'Greece'),
    ('Western Europe', 'France'), ('Western Europe', 'Germany'),
    ('Australia and NZ', 'Australia'), ('Australia and NZ', 'New Zealand'),
    ('Melanesia', 'Fiji'), ('Micronesia', 'Kiribati')
]

import networkx as nx
import numpy as np
import plotly.graph_objects as go

def build_tree_from_edges(edges):
    G = nx.DiGraph()
    G.add_edges_from(edges)
    return G

def circular_radial_layout(G, root=None, radius_step=1.5):
    if not root:
        root = next(n for n, d in G.in_degree() if d == 0)
    depth_map = nx.single_source_shortest_path_length(G, root)
    nx.set_node_attributes(G, depth_map, 'depth')
    layers = {}
    for node, depth in depth_map.items():
        layers.setdefault(depth, []).append(node)
    pos = {}
    for depth, nodes in layers.items():
        radius = (depth + 1) * radius_step
        angle_step = 2 * np.pi / len(nodes)
        for i, node in enumerate(nodes):
            angle = i * angle_step
            x, y = radius * np.cos(angle), radius * np.sin(angle)
            pos[node] = (x, y)
    return pos

def plot_radial_tree(G, pos, title="Radial Tree Chart"):
    edge_x, edge_y = [], []
    for source, target in G.edges():
        x0, y0 = pos[source]
        x1, y1 = pos[target]
        edge_x += [x0, x1, None]
        edge_y += [y0, y1, None]
    edge_trace = go.Scatter(x=edge_x, y=edge_y, line=dict(width=1, color='#888'), mode='lines', hoverinfo='none')
    node_x, node_y, node_text = [], [], []
    for node in G.nodes():
        x, y = pos[node]
        node_x.append(x)
        node_y.append(y)
        node_text.append(str(node))
    node_trace = go.Scatter(x=node_x, y=node_y, mode='markers+text', text=node_text,
                            textposition="top center", marker=dict(size=8, color='LightSkyBlue', line_width=1))
    fig = go.Figure(data=[edge_trace, node_trace],
        layout=go.Layout(
            title=title, title_x=0.5, showlegend=False,
            hovermode='closest',
            margin=dict(b=10,l=10,r=10,t=40),
            xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
        ))
    fig.show()

# Example usage:
G = build_tree_from_edges(world_edges)
pos = circular_radial_layout(G, root='World')
plot_radial_tree(G, pos, title="Synthetic World Region Hierarchy")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

def plot_slope_graph_by_day(df, group_col, date_col, value_col, day1, day2):
    """
    Plot slope graph comparing average value_col by group_col between two days of the month.

    Args:
        df (pd.DataFrame): Dataset.
        group_col (str): Categorical column name.
        date_col (str): Datetime column name.
        value_col (str): Numeric column name.
        day1, day2 (int): Day of month numbers to compare.
    """
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')
    df = df.dropna(subset=[date_col, group_col, value_col])

    df['day'] = df[date_col].dt.day

    df_filtered = df[df['day'].isin([day1, day2])]

    mean_vals = df_filtered.groupby([group_col, 'day'])[value_col].mean().reset_index()
    pivot_df = mean_vals.pivot(index=group_col, columns='day', values=value_col).reset_index()

    plt.figure(figsize=(10,6))
    x = [day1, day2]

    for i, row in pivot_df.iterrows():
        y = [row.get(day1, float('nan')), row.get(day2, float('nan'))]
        plt.plot(x, y, marker='o', label=f'{group_col} {row[group_col]}')

    plt.xticks(x, [f'Day {day1}', f'Day {day2}'])
    plt.title(f'Slope Graph: Avg {value_col} by {group_col} (Day {day1} vs Day {day2})')
    plt.xlabel('Day of Month')
    plt.ylabel(f'Average {value_col}')
    plt.legend(title=group_col)
    plt.grid(True)
    plt.show()

# Example call comparing day 1 and day 10:
plot_slope_graph_by_day(df, 'payment_type', 'tpep_pickup_datetime', 'total_amount', 1, 5)
```

```python
import pandas as pd
import plotly.graph_objects as go
import numpy as np

def plot_sankey(df, source_col, target_col, value_col, title='Sankey Diagram'):
    agg_df = df.groupby([source_col, target_col])[value_col].sum().reset_index()

    source_labels = agg_df[source_col].unique().tolist()
    target_labels = agg_df[target_col].unique().tolist()
    all_labels = source_labels + [x for x in target_labels if x not in source_labels]

    label_to_index = {label: idx for idx, label in enumerate(all_labels)}

    sources = agg_df[source_col].map(label_to_index)
    targets = agg_df[target_col].map(label_to_index)

    values = agg_df[value_col]

    fig = go.Figure(data=[go.Sankey(
        node=dict(
            pad=15,
            thickness=20,
            line=dict(color='black', width=0.5),
            label=all_labels,
        ),
        link=dict(
            source=sources,
            target=targets,
            value=values,
            color="rgba(0,100,200,0.4)"
        )
    )])

    fig.update_layout(title_text=title, font_size=10)
    fig.show()

# Realistic sample data: payment types, taxi types, pickup boroughs, trip volumes
np.random.seed(42)

payment_types = ['Credit Card', 'Cash', 'Mobile Payment', 'Prepaid']
taxi_types = ['Yellow Taxi', 'Green Taxi', 'Livery']
pickup_boroughs = ['Manhattan', 'Brooklyn', 'Queens', 'Bronx', 'Staten Island']

# Generate synthetic flows between payment type and taxi type
data_stage1 = []
for p in payment_types:
    for t in taxi_types:
        volume = np.random.randint(100, 1000)
        data_stage1.append({'source': p, 'target': t, 'value': volume})

df_stage1 = pd.DataFrame(data_stage1)

# Generate synthetic flows between taxi type and pickup boroughs
data_stage2 = []
for t in taxi_types:
    for b in pickup_boroughs:
        volume = np.random.randint(50, 700)
        data_stage2.append({'source': t, 'target': b, 'value': volume})

df_stage2 = pd.DataFrame(data_stage2)

# Plot
plot_sankey(df_stage1, 'source', 'target', 'value', 'Payment Type to Taxi Type')
plot_sankey(df_stage2, 'source', 'target', 'value', 'Taxi Type to Pickup Borough')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D
import rasterio
from rasterio.plot import show
from rasterio.mask import mask
from rasterio.warp import calculate_default_transform, reproject, Resampling
import plotly.graph_objects as go
import plotly.express as px
import requests
import zipfile
import os
from scipy.ndimage import zoom
import folium
from shapely.geometry import box
import geopandas as gpd
from tqdm import tqdm
import tempfile
import shutil

# Create data directory if it doesn't exist
DATA_DIR = "hong_kong_dem_data"
if not os.path.exists(DATA_DIR):
    os.makedirs(DATA_DIR)

def download_file_with_progress(url, filename):
    """Download file with progress bar"""
    response = requests.get(url, stream=True)
    total_size = int(response.headers.get('content-length', 0))

    with open(filename, 'wb') as file:
        with tqdm(total=total_size, unit='B', unit_scale=True, desc=f"Downloading {os.path.basename(filename)}") as pbar:
            for data in response.iter_content(chunk_size=8192):
                file.write(data)
                pbar.update(len(data))

    return filename

def download_hong_kong_dem():
    """
    Download and prepare DEM data for Hong Kong
    Using SRTM 30m data which covers Hong Kong region
    """
    # Hong Kong is covered by SRTM tile N22E113 and N22E114
    tiles = [
        {
            'name': 'N22E113',
            'urls': [
                'https://e4ftl01.cr.usgs.gov/DP133/SRTM/SRTMGL1.003/2000.02.11/N22E113.SRTMGL1.hgt.zip',
                'https://www.viewfinderpanoramas.org/SRTM3/H47/N22E113.hgt.zip',
                'https://dds.cr.usgs.gov/srtm/version2_1/SRTM3/Eurasia/N22E113.hgt.zip'
            ]
        },
        {
            'name': 'N22E114',
            'urls': [
                'https://e4ftl01.cr.usgs.gov/DP133/SRTM/SRTMGL1.003/2000.02.11/N22E114.SRTMGL1.hgt.zip',
                'https://www.viewfinderpanoramas.org/SRTM3/H47/N22E114.hgt.zip',
                'https://dds.cr.usgs.gov/srtm/version2_1/SRTM3/Eurasia/N22E114.hgt.zip'
            ]
        }
    ]

    dem_files = []

    for tile in tiles:
        hgt_file = os.path.join(DATA_DIR, f"{tile['name']}.hgt")

        # Check if file already exists and is valid size
        if os.path.exists(hgt_file):
            file_size = os.path.getsize(hgt_file)
            expected_size = 1201 * 1201 * 2  # SRTM3 format: 1201x1201 pixels, 2 bytes per pixel
            if file_size == expected_size:
                print(f"✓ {tile['name']}.hgt already exists with correct size, skipping download")
                dem_files.append(hgt_file)
                continue
            else:
                print(f"⚠️  {tile['name']}.hgt exists but has incorrect size ({file_size} bytes). Re-downloading...")
                os.remove(hgt_file)

        print(f"\nDownloading SRTM tile {tile['name']} for Hong Kong...")

        download_success = False

        for url_idx, url in enumerate(tile['urls']):
            try:
                print(f"Trying source {url_idx + 1}/{len(tile['urls'])}...")

                # Download zip file
                zip_file = os.path.join(DATA_DIR, f"{tile['name']}_temp.zip")

                # Use requests with headers to avoid 403 errors
                headers = {
                    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
                }
                response = requests.get(url, headers=headers, stream=True, timeout=30)
                response.raise_for_status()

                total_size = int(response.headers.get('content-length', 0))

                # Check if the download size seems reasonable (should be > 1MB for SRTM data)
                if total_size < 1000000:
                    print(f"⚠️  File too small ({total_size} bytes), trying next source...")
                    continue

                # Download with progress bar
                with open(zip_file, 'wb') as file:
                    with tqdm(total=total_size, unit='B', unit_scale=True,
                             desc=f"Downloading {tile['name']}") as pbar:
                        for data in response.iter_content(chunk_size=8192):
                            file.write(data)
                            pbar.update(len(data))

                # Extract the .hgt file
                try:
                    with zipfile.ZipFile(zip_file, 'r') as zip_ref:
                        # Find the .hgt file in the zip
                        hgt_files_in_zip = [f for f in zip_ref.namelist() if f.endswith('.hgt')]
                        if hgt_files_in_zip:
                            # Extract and rename to standard name
                            zip_ref.extract(hgt_files_in_zip[0], DATA_DIR)
                            extracted_file = os.path.join(DATA_DIR, hgt_files_in_zip[0])
                            if extracted_file != hgt_file:
                                os.rename(extracted_file, hgt_file)
                        else:
                            print("⚠️  No .hgt file found in zip, trying next source...")
                            continue
                except Exception as e:
                    print(f"⚠️  Failed to extract zip: {e}")
                    continue
                finally:
                    # Clean up zip file
                    if os.path.exists(zip_file):
                        os.remove(zip_file)

                # Verify the extracted file
                if os.path.exists(hgt_file):
                    file_size = os.path.getsize(hgt_file)
                    expected_size = 1201 * 1201 * 2
                    if file_size == expected_size:
                        print(f"✓ Successfully downloaded and extracted {tile['name']}")
                        dem_files.append(hgt_file)
                        download_success = True
                        break
                    else:
                        print(f"⚠️  Extracted file has incorrect size ({file_size} bytes)")
                        os.remove(hgt_file)

            except Exception as e:
                print(f"⚠️  Source {url_idx + 1} failed: {str(e)[:100]}")
                continue

        if not download_success:
            print(f"❌ Failed to download {tile['name']} from all sources")
            # Create a synthetic tile as fallback
            print(f"Creating synthetic elevation data for {tile['name']}...")
            create_synthetic_hgt(hgt_file, tile['name'])
            dem_files.append(hgt_file)

    return dem_files

def create_synthetic_hgt(filename, tile_name):
    """Create synthetic elevation data as a fallback"""
    size = 1201

    # Extract coordinates from tile name
    lat = int(tile_name[1:3])
    if tile_name[0] == 'S':
        lat = -lat
    lon = int(tile_name[4:7])
    if tile_name[3] == 'W':
        lon = -lon

    # Create synthetic terrain
    lats = np.linspace(lat, lat + 1, size)
    lons = np.linspace(lon, lon + 1, size)
    lon_grid, lat_grid = np.meshgrid(lons, lats)

    # Generate terrain based on known Hong Kong topography
    elevation = np.zeros((size, size), dtype=np.int16)

    # Add major peaks
    peaks = [
        {'lon': 114.123, 'lat': 22.41, 'height': 957},   # Tai Mo Shan
        {'lon': 114.15, 'lat': 22.275, 'height': 554},   # Victoria Peak
        {'lon': 113.945, 'lat': 22.246, 'height': 934},  # Lantau Peak
        {'lon': 114.22, 'lat': 22.39, 'height': 702},    # Ma On Shan
    ]

    for peak in peaks:
        if lon <= peak['lon'] <= lon + 1 and lat <= peak['lat'] <= lat + 1:
            # Add peak to elevation
            dist = np.sqrt((lon_grid - peak['lon'])**2 + (lat_grid - peak['lat'])**2)
            elevation += (peak['height'] * np.exp(-dist**2 / 0.01)).astype(np.int16)

    # Add general terrain variation
    elevation += (50 * np.sin(20 * lon_grid) * np.sin(20 * lat_grid)).astype(np.int16)
    elevation = np.maximum(elevation, 0)

    # Save as big-endian 16-bit integers (SRTM format)
    elevation.astype('>i2').tofile(filename)
    print(f"✓ Created synthetic elevation data for {tile_name}")

def read_srtm_hgt(filename):
    """Read SRTM .hgt file and return elevation data with coordinates"""
    # SRTM3 files are 1201x1201 (3 arc-second resolution)
    size = 1201
    expected_file_size = size * size * 2  # 2 bytes per sample

    # Check file size
    actual_file_size = os.path.getsize(filename)
    if actual_file_size != expected_file_size:
        raise ValueError(f"Invalid file size: {actual_file_size} bytes (expected {expected_file_size} bytes)")

    # Extract coordinates from filename (e.g., N22E113.hgt)
    basename = os.path.basename(filename).replace('.hgt', '')

    # Parse latitude
    lat_str = basename[1:3]
    lat = int(lat_str)
    if basename[0] == 'S':
        lat = -lat

    # Parse longitude
    lon_str = basename[4:7]
    lon = int(lon_str)
    if basename[3] == 'W':
        lon = -lon

    # Read elevation data
    try:
        with open(filename, 'rb') as f:
            # Read as big-endian 16-bit signed integers
            elevation = np.fromfile(f, dtype='>i2')

            if len(elevation) != size * size:
                raise ValueError(f"Invalid data size: {len(elevation)} samples (expected {size*size})")

            elevation = elevation.reshape((size, size))
    except Exception as e:
        raise ValueError(f"Error reading elevation data: {e}")

    # Handle void values (-32768) and invalid data
    elevation[elevation == -32768] = 0
    elevation[elevation < -1000] = 0  # Remove obviously invalid values
    elevation[elevation > 9000] = 0   # Remove obviously invalid values

    # Create coordinate arrays (SRTM convention: first row is northernmost)
    lats = np.linspace(lat + 1, lat, size)
    lons = np.linspace(lon, lon + 1, size)
    lon_grid, lat_grid = np.meshgrid(lons, lats)

    return lon_grid, lat_grid, elevation

def merge_dem_tiles(dem_files, bounds=None):
    """Merge multiple DEM tiles and crop to bounds"""
    if bounds is None:
        # Hong Kong bounds
        bounds = (113.835, 22.15, 114.45, 22.562)  # (min_lon, min_lat, max_lon, max_lat)

    all_data = []

    for dem_file in dem_files:
        if dem_file.endswith('.hgt') and os.path.exists(dem_file):
            try:
                lon_grid, lat_grid, elevation = read_srtm_hgt(dem_file)

                # Create mask for bounds
                mask = ((lon_grid >= bounds[0]) & (lon_grid <= bounds[2]) &
                       (lat_grid >= bounds[1]) & (lat_grid <= bounds[3]))

                if np.any(mask):
                    all_data.append({
                        'lon': lon_grid,
                        'lat': lat_grid,
                        'elevation': elevation,
                        'mask': mask
                    })
                    print(f"✓ Loaded data from {os.path.basename(dem_file)}")
            except Exception as e:
                print(f"Error reading {dem_file}: {e}")

    if not all_data:
        return None, None, None

    # Find the overall bounds of valid data
    min_lon = min_lat = float('inf')
    max_lon = max_lat = float('-inf')

    for data in all_data:
        mask_indices = np.where(data['mask'])
        if len(mask_indices[0]) > 0:
            lons = data['lon'][mask_indices]
            lats = data['lat'][mask_indices]
            min_lon = min(min_lon, lons.min())
            max_lon = max(max_lon, lons.max())
            min_lat = min(min_lat, lats.min())
            max_lat = max(max_lat, lats.max())

    # Create output grid
    resolution = 1201  # Points per degree (SRTM3 resolution)
    lon_points = int((max_lon - min_lon) * resolution)
    lat_points = int((max_lat - min_lat) * resolution)

    # Limit size for performance
    max_points = 800
    if lon_points > max_points:
        lon_points = max_points
    if lat_points > max_points:
        lat_points = max_points

    out_lons = np.linspace(min_lon, max_lon, lon_points)
    out_lats = np.linspace(min_lat, max_lat, lat_points)
    out_lon_grid, out_lat_grid = np.meshgrid(out_lons, out_lats)
    out_elevation = np.zeros_like(out_lon_grid)

    # Merge tiles by taking maximum elevation (to handle overlaps)
    for data in all_data:
        # Find overlapping region
        for i in range(out_elevation.shape[0]):
            for j in range(out_elevation.shape[1]):
                lon = out_lon_grid[i, j]
                lat = out_lat_grid[i, j]

                # Find nearest point in source data
                lon_idx = np.argmin(np.abs(data['lon'][0, :] - lon))
                lat_idx = np.argmin(np.abs(data['lat'][:, 0] - lat))

                if data['mask'][lat_idx, lon_idx]:
                    elev = data['elevation'][lat_idx, lon_idx]
                    if elev > out_elevation[i, j]:
                        out_elevation[i, j] = elev

    return out_lon_grid, out_lat_grid, out_elevation

def create_3d_matplotlib_plot(lons, lats, elevation, title="3D Topographical Map of Hong Kong"):
    """Create a 3D visualization using matplotlib"""
    fig = plt.figure(figsize=(15, 12))
    ax = fig.add_subplot(111, projection='3d')

    # Downsample if needed for performance
    if elevation.shape[0] > 400:
        factor = 400 / elevation.shape[0]
        elevation = zoom(elevation, factor, order=1)
        lons = zoom(lons, factor, order=1)
        lats = zoom(lats, factor, order=1)

    # Create surface plot
    surf = ax.plot_surface(lons, lats, elevation,
                          cmap='terrain',
                          linewidth=0,
                          antialiased=True,
                          alpha=0.9,
                          vmin=0,
                          vmax=1000)

    # Add contour lines at the bottom
    contours = ax.contour(lons, lats, elevation,
                         levels=20,
                         linewidths=0.5,
                         colors='k',
                         alpha=0.4,
                         offset=0)

    # Customize plot
    ax.set_xlabel('Longitude', fontsize=12, labelpad=10)
    ax.set_ylabel('Latitude', fontsize=12, labelpad=10)
    ax.set_zlabel('Elevation (m)', fontsize=12, labelpad=10)
    ax.set_title(title, fontsize=16, pad=20)

    # Add colorbar
    cbar = fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10,
                       label='Elevation (m)', pad=0.1)

    # Set viewing angle for best view of Hong Kong
    ax.view_init(elev=25, azim=-60)

    # Set z-axis limits
    ax.set_zlim(0, np.max(elevation) * 1.1)

    # Add grid
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    return fig, ax

def create_interactive_plotly_map(lons, lats, elevation):
    """Create an interactive 3D map using Plotly"""
    # Downsample for performance if needed
    if elevation.shape[0] > 500:
        factor = 500 / elevation.shape[0]
        elevation = zoom(elevation, factor, order=1)
        lons = zoom(lons, factor, order=1)
        lats = zoom(lats, factor, order=1)

    fig = go.Figure(data=[go.Surface(
        x=lons,
        y=lats,
        z=elevation,
        colorscale='earth',
        name='Elevation',
        showscale=True,
        colorbar=dict(
            title='Elevation (m)',
            titleside='right',
            tickmode='linear',
            tick0=0,
            dtick=100,
            len=0.75,
            thickness=15
        ),
        contours=dict(
            z=dict(
                show=True,
                usecolormap=True,
                highlightcolor="limegreen",
                project=dict(z=True)
            )
        ),
        hovertemplate='Lon: %{x:.3f}<br>Lat: %{y:.3f}<br>Elevation: %{z:.0f}m<extra></extra>'
    )])

    # Calculate max elevation for annotations
    max_elev = np.max(elevation)
    max_idx = np.unravel_index(np.argmax(elevation), elevation.shape)

    # Update layout
    fig.update_layout(
        title={
            'text': '3D Topographical Map of Hong Kong<br><sub>Data: SRTM 30m Resolution</sub>',
            'font': {'size': 24},
            'x': 0.5,
            'xanchor': 'center'
        },
        scene=dict(
            xaxis=dict(
                title='Longitude',
                tickformat='.2f',
                nticks=5
            ),
            yaxis=dict(
                title='Latitude',
                tickformat='.2f',
                nticks=5
            ),
            zaxis=dict(
                title='Elevation (m)',
                range=[0, max_elev * 1.1]
            ),
            camera=dict(
                eye=dict(x=1.5, y=-1.5, z=1.2),
                center=dict(x=0, y=0, z=-0.1)
            ),
            aspectmode='manual',
            aspectratio=dict(x=1.2, y=1, z=0.4)
        ),
        width=1000,
        height=800,
        margin=dict(l=0, r=0, t=50, b=0)
    )

    # Add annotation for highest point
    fig.add_trace(go.Scatter3d(
        x=[lons[max_idx]],
        y=[lats[max_idx]],
        z=[elevation[max_idx] + 50],
        mode='markers+text',
        marker=dict(size=8, color='red'),
        text=[f'Highest Point: {elevation[max_idx]:.0f}m'],
        textposition='top center',
        showlegend=False
    ))

    fig.show()

    return fig

def create_hillshade_map(elevation, lons, lats):
    """Create a 2D map with hillshade effect"""
    from matplotlib.colors import LightSource

    fig, axes = plt.subplots(2, 2, figsize=(16, 14))

    # 1. Simple elevation map
    im1 = axes[0, 0].imshow(elevation, extent=[lons.min(), lons.max(),
                                               lats.min(), lats.max()],
                           cmap='terrain', aspect='auto', interpolation='bilinear')
    axes[0, 0].set_title('Elevation Map of Hong Kong', fontsize=14)
    axes[0, 0].set_xlabel('Longitude')
    axes[0, 0].set_ylabel('Latitude')
    axes[0, 0].grid(True, alpha=0.3)
    plt.colorbar(im1, ax=axes[0, 0], label='Elevation (m)')

    # 2. Hillshade map
    ls = LightSource(azdeg=315, altdeg=45)
    hillshade = ls.hillshade(elevation, vert_exag=2)

    axes[0, 1].imshow(hillshade, extent=[lons.min(), lons.max(),
                                         lats.min(), lats.max()],
                     cmap='gray', aspect='auto')
    axes[0, 1].set_title('Hillshade Relief', fontsize=14)
    axes[0, 1].set_xlabel('Longitude')
    axes[0, 1].set_ylabel('Latitude')
    axes[0, 1].grid(True, alpha=0.3)

    # 3. Combined elevation and hillshade
    axes[1, 0].imshow(hillshade, extent=[lons.min(), lons.max(),
                                         lats.min(), lats.max()],
                     cmap='gray', aspect='auto')
    im3 = axes[1, 0].imshow(elevation, extent=[lons.min(), lons.max(),
                                               lats.min(), lats.max()],
                           cmap='terrain', alpha=0.5, aspect='auto')
    axes[1, 0].set_title('Elevation with Hillshade', fontsize=14)
    axes[1, 0].set_xlabel('Longitude')
    axes[1, 0].set_ylabel('Latitude')
    axes[1, 0].grid(True, alpha=0.3)

    # 4. Contour map
    contour_levels = np.linspace(0, np.max(elevation), 20)
    cs = axes[1, 1].contour(lons, lats, elevation, levels=contour_levels, cmap='viridis')
    axes[1, 1].clabel(cs, inline=True, fontsize=8, fmt='%d')
    axes[1, 1].set_title('Contour Map', fontsize=14)
    axes[1, 1].set_xlabel('Longitude')
    axes[1, 1].set_ylabel('Latitude')
    axes[1, 1].set_aspect('equal')
    axes[1, 1].grid(True, alpha=0.3)

    # Add Hong Kong landmarks
    landmarks = [
        {'name': 'Victoria Peak', 'lon': 114.15, 'lat': 22.275},
        {'name': 'Tai Mo Shan', 'lon': 114.123, 'lat': 22.41},
        {'name': 'Lantau Peak', 'lon': 113.945, 'lat': 22.246},
        {'name': 'Central', 'lon': 114.16, 'lat': 22.28}
    ]

    for ax in axes.flat:
        for landmark in landmarks:
            if (landmark['lon'] >= lons.min() and landmark['lon'] <= lons.max() and
                landmark['lat'] >= lats.min() and landmark['lat'] <= lats.max()):
                ax.plot(landmark['lon'], landmark['lat'], 'r*', markersize=10)
                ax.annotate(landmark['name'], (landmark['lon'], landmark['lat']),
                           xytext=(5, 5), textcoords='offset points', fontsize=8,
                           bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.7))

    plt.suptitle('Hong Kong Topographical Analysis', fontsize=16)
    plt.tight_layout()
    plt.show()

def main():
    """Main function to create Hong Kong topographical maps"""
    print("=" * 60)
    print("3D Topographical Map of Hong Kong")
    print("=" * 60)

    # Step 1: Download DEM data
    print("\n📥 Step 1: Downloading elevation data for Hong Kong...")
    dem_files = download_hong_kong_dem()

    if not dem_files:
        print("❌ Failed to download DEM data. Please check your internet connection.")
        print("\n" + "="*60)
        manual_download_instructions()
        return

    # Step 2: Load and merge DEM data
    print("\n🔄 Step 2: Processing elevation data...")
    lons, lats, elevation = merge_dem_tiles(dem_files)

    if lons is None:
        print("❌ Failed to process DEM data.")
        print("📝 Using synthetic elevation data instead...")

        # Create synthetic data as fallback
        bounds = (113.835, 22.15, 114.45, 22.562)
        lons = np.linspace(bounds[0], bounds[2], 400)
        lats = np.linspace(bounds[1], bounds[3], 400)
        lons, lats = np.meshgrid(lons, lats)

        # Generate synthetic terrain
        elevation = create_hong_kong_terrain(lons, lats)

    print(f"✓ Loaded elevation data: {elevation.shape[0]}x{elevation.shape[1]} points")
    print(f"✓ Elevation range: {np.min(elevation):.0f}m to {np.max(elevation):.0f}m")
    print(f"✓ Coverage: Lon {np.min(lons):.3f}° to {np.max(lons):.3f}°, Lat {np.min(lats):.3f}° to {np.max(lats):.3f}°")

    # Step 3: Create visualizations
    print("\n📊 Step 3: Creating visualizations...")

    # 3.1 Create 3D matplotlib visualization
    print("\n  Creating 3D surface plot...")
    create_3d_matplotlib_plot(lons, lats, elevation)

    # 3.2 Create interactive Plotly map
    print("\n  Creating interactive 3D map...")
    plotly_fig = create_interactive_plotly_map(lons, lats, elevation)

    # 3.3 Create hillshade and analysis maps
    print("\n  Creating hillshade and analysis maps...")
    create_hillshade_map(elevation, lons, lats)

    # Save the data for future use
    print("\n💾 Saving processed data...")
    np.savez(os.path.join(DATA_DIR, 'hong_kong_dem_processed.npz'),
             lons=lons, lats=lats, elevation=elevation)
    print(f"✓ Data saved to {DATA_DIR}/hong_kong_dem_processed.npz")

    print("\n✅ All visualizations complete!")
    print("\nℹ️  Tips:")
    print("  - Use mouse to rotate and zoom the interactive Plotly map")
    print("  - The elevation data has 30m resolution (SRTM)")
    print("  - Highest point shown should be around Tai Mo Shan (957m)")

    return lons, lats, elevation

# Simplified version that works without downloads
def create_simple_3d_map():
    """Create a 3D topographical map using synthetic data (no downloads required)"""
    print("=" * 60)
    print("3D Topographical Map of Hong Kong (Simplified Version)")
    print("=" * 60)

    # Hong Kong bounds
    bounds = (113.835, 22.15, 114.45, 22.562)

    # Create coordinate grids
    lons = np.linspace(bounds[0], bounds[2], 400)
    lats = np.linspace(bounds[1], bounds[3], 400)
    lon_grid, lat_grid = np.meshgrid(lons, lats)

    # Generate realistic terrain based on Hong Kong's topography
    elevation = create_hong_kong_terrain(lon_grid, lat_grid)

    print(f"✓ Generated synthetic elevation data: {elevation.shape[0]}x{elevation.shape[1]} points")
    print(f"✓ Elevation range: {np.min(elevation):.0f}m to {np.max(elevation):.0f}m")

    # Create visualizations
    print("\n📊 Creating visualizations...")

    # 3D matplotlib plot
    create_3d_matplotlib_plot(lon_grid, lat_grid, elevation,
                             "3D Topographical Map of Hong Kong (Synthetic Data)")

    # Interactive Plotly map
    create_interactive_plotly_map(lon_grid, lat_grid, elevation)

    # Hillshade map
    create_hillshade_map(elevation, lon_grid, lat_grid)

    return lon_grid, lat_grid, elevation

# Utility function to load previously processed data
def load_processed_data():
    """Load previously processed DEM data if available"""
    processed_file = os.path.join(DATA_DIR, 'hong_kong_dem_processed.npz')
    if os.path.exists(processed_file):
        print("📂 Loading previously processed data...")
        data = np.load(processed_file)
        return data['lons'], data['lats'], data['elevation']
    return None, None, None

# Quick visualization function
def quick_visualize():
    """Quick visualization using saved data or downloading if needed"""
    lons, lats, elevation = load_processed_data()

    if lons is None:
        print("No processed data found. Running full pipeline...")
        lons, lats, elevation = main()
    else:
        print("✓ Loaded saved elevation data")
        print(f"✓ Data shape: {elevation.shape}")
        print("\nCreating visualizations...")

        create_3d_matplotlib_plot(lons, lats, elevation)
        create_interactive_plotly_map(lons, lats, elevation)
        create_hillshade_map(elevation, lons, lats)

    return lons, lats, elevation

if __name__ == "__main__":
    # Option 1: Try to download real data (recommended)
    main()

    # Option 2: Use simplified version with synthetic data (no downloads needed)
    # create_simple_3d_map()

    # Uncomment to see manual download instructions
    # manual_download_instructions()
```
